```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Generate dummy numerical data
np.random.seed(42) # for reproducibility
dummy_data = {
    'value': np.random.normal(loc=50, scale=15, size=1000) # Normally distributed data
}
df_dummy = pd.DataFrame(dummy_data)

display(df_dummy.head())
```

### Histogram

Histogram adalah representasi grafis yang mengorganisasikan sekelompok titik data ke dalam rentang yang ditentukan pengguna. Penampilannya mirip dengan grafik batang, tetapi ia memadatkan kumpulan data yang besar menjadi visual yang mudah diinterpretasikan. Ini membantu untuk memvisualisasikan distribusi variabel numerik.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_dummy['value'], bins=30, kde=True)
plt.title('Histogram')
plt.xlabel('Nilai')
plt.ylabel('Frekuensi')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

### Cara Membaca Histogram

Untuk membaca histogram:

1.  **Sumbu X (Horizontal):** Menunjukkan rentang nilai dari variabel numerik yang sedang dianalisis (misalnya, 'Nilai' dalam contoh ini). Rentang ini dibagi menjadi interval yang disebut 'bins' atau 'kelas'.
2.  **Sumbu Y (Vertikal):** Menunjukkan frekuensi atau jumlah data yang jatuh ke dalam setiap 'bin'. Semakin tinggi batang, semakin banyak data yang berada dalam rentang nilai tersebut.
3.  **Bentuk Distribusi:** Perhatikan bentuk keseluruhan histogram:
    *   **Simetris/Distribusi Normal:** Jika batangnya membentuk pola seperti lonceng (tinggi di tengah, rendah di ujung), seperti contoh di atas.
    *   **Miring Kanan (Right-Skewed):** Jika 'ekor' grafik lebih panjang ke arah kanan (nilai-nilai yang lebih tinggi), artinya sebagian besar data terkonsentrasi di nilai-nilai rendah.
    *   **Miring Kiri (Left-Skewed):** Jika 'ekor' grafik lebih panjang ke arah kiri (nilai-nilai yang lebih rendah), artinya sebagian besar data terkonsentrasi di nilai-nilai tinggi.
    *   **Bimodal/Multimodal:** Jika ada dua atau lebih puncak yang jelas, menunjukkan adanya beberapa kelompok berbeda dalam data.
4.  **Penyebaran:** Seberapa lebar distribusi tersebut. Distribusi yang lebar menunjukkan variasi data yang besar.
5.  **Pencilan (Outlier):** Meskipun histogram tidak sejelas boxplot dalam menunjukkan outlier, nilai-nilai yang sangat ekstrem dan jauh dari sebagian besar data mungkin muncul sebagai batang yang sangat kecil di ujung grafik.

### Boxplot

Boxplot adalah cara standar untuk menampilkan distribusi data berdasarkan ringkasan lima angka: minimum, kuartil pertama (Q1), median, kuartil ketiga (Q3), dan maksimum. Ini juga dapat menunjukkan outlier dalam data.

```python
plt.figure(figsize=(10, 6))
sns.boxplot(y=df_dummy['value'])
plt.title('Boxplot')
plt.ylabel('Nilai')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

### Cara Membaca Boxplot

Untuk membaca boxplot, perhatikan lima komponen utamanya:

1.  **Garis Tengah dalam Kotak (Median):** Garis ini membagi kotak menjadi dua bagian dan mewakili median (nilai tengah) dari kumpulan data. 50% data berada di atas garis ini dan 50% di bawahnya.
2.  **Tepi Bawah Kotak (Kuartil Pertama, Q1):** Ini adalah nilai di mana 25% data berada di bawahnya. Ini juga dikenal sebagai persentil ke-25.
3.  **Tepi Atas Kotak (Kuartil Ketiga, Q3):** Ini adalah nilai di mana 75% data berada di bawahnya. Ini juga dikenal sebagai persentil ke-75.
4.  **Kotak (Interquartile Range - IQR):** Area antara Q1 dan Q3. Kotak ini mencakup 50% data tengah dan menunjukkan seberapa tersebar data di sekitar median. Semakin panjang kotak, semakin besar penyebaran data di bagian tengah.
5.  **Kumisan (Whiskers):** Garis yang memanjang dari kotak. Kumisan biasanya menunjukkan:
    *   **Kumisan Bawah:** Nilai terendah dalam data yang *bukan* outlier, dihitung sebagai Q1 - 1.5 * IQR.
    *   **Kumisan Atas:** Nilai tertinggi dalam data yang *bukan* outlier, dihitung sebagai Q3 + 1.5 * IQR.
6.  **Titik-titik di Luar Kumisan (Outliers):** Titik-titik individual di luar kumisan menunjukkan nilai-nilai yang dianggap outlier. Ini adalah nilai yang secara signifikan berbeda dari sebagian besar kumpulan data dan bisa mengindikasikan kesalahan data atau kejadian yang tidak biasa.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Generate dummy categorical data
np.random.seed(42) # for reproducibility

dummy_categories = np.random.choice(['A', 'B', 'C', 'D'], size=1000, p=[0.25, 0.35, 0.20, 0.20])
dummy_values = np.random.randint(10, 100, size=1000) # numerical values for aggregation

df_categorical = pd.DataFrame({
    'Kategori': dummy_categories,
    'Nilai_Numerik': dummy_values
})

display(df_categorical.head())
```

### 1. Grafik Batang Frekuensi


Grafik batang frekuensi adalah jenis grafik batang yang digunakan untuk menampilkan distribusi data kategorikal. Setiap batang mewakili satu kategori, dan tinggi batangnya menunjukkan jumlah (frekuensi) kemunculan kategori tersebut dalam kumpulan data. Grafik ini sangat berguna untuk melihat kategori mana yang paling sering atau paling jarang muncul dalam data Anda.

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Kategori', data=df_categorical, palette='viridis')
plt.title('Grafik Batang Frekuensi Kategori')
plt.xlabel('Kategori')
plt.ylabel('Frekuensi')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

#### Cara Membaca Grafik Batang Frekuensi

Untuk membaca grafik batang frekuensi:

1.  **Sumbu X (Horizontal):** Mewakili kategori-kategori yang berbeda dalam data (misalnya, 'A', 'B', 'C', 'D').
2.  **Sumbu Y (Vertikal):** Menunjukkan jumlah atau frekuensi kemunculan setiap kategori.
3.  **Tinggi Batang:** Semakin tinggi batang untuk suatu kategori, semakin banyak kemunculan kategori tersebut dalam data.
4.  **Perbandingan:** Anda dapat dengan mudah membandingkan popularitas atau keberadaan relatif setiap kategori dengan melihat tinggi batangnya. Kategori dengan batang tertinggi adalah yang paling sering, sedangkan yang terendah adalah yang paling jarang.

### 2. Grafik Batang Nilai Agregat


Grafik batang nilai agregat digunakan untuk menampilkan ringkasan numerik (seperti jumlah, rata-rata, median, dll.) dari suatu variabel kuantitatif untuk setiap kategori dari variabel kategorikal. Misalnya, Anda ingin melihat rata-rata 'Nilai_Numerik' untuk setiap 'Kategori'. Ini membantu dalam membandingkan karakteristik numerik antar kategori.

```python
# Menghitung rata-rata Nilai_Numerik per Kategori
df_agg_mean = df_categorical.groupby('Kategori')['Nilai_Numerik'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='Kategori', y='Nilai_Numerik', data=df_agg_mean, palette='magma')
plt.title('Grafik Batang Rata-rata Nilai Numerik per Kategori')
plt.xlabel('Kategori')
plt.ylabel('Rata-rata Nilai Numerik')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

#### Cara Membaca Grafik Batang Nilai Agregat

Untuk membaca grafik batang nilai agregat:

1.  **Sumbu X (Horizontal):** Mewakili kategori-kategori yang berbeda.
2.  **Sumbu Y (Vertikal):** Menunjukkan nilai agregat (dalam contoh ini, rata-rata 'Nilai_Numerik') untuk setiap kategori.
3.  **Tinggi Batang:** Tinggi setiap batang menunjukkan nilai agregat dari variabel numerik untuk kategori yang sesuai. Misalnya, jika Anda menghitung rata-rata penjualan per produk, tinggi batang akan menunjukkan rata-rata penjualan untuk setiap produk.
4.  **Perbandingan:** Anda dapat dengan mudah membandingkan nilai agregat antar kategori. Kategori dengan batang tertinggi memiliki nilai agregat tertinggi, dan sebaliknya. Ini membantu mengidentifikasi kategori mana yang berkinerja lebih baik atau berbeda secara signifikan dalam metrik yang diukur.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Generate dummy data for scatter plot
np.random.seed(42) # for reproducibility

# Create a variable (X)
X = np.random.rand(100) * 10 # Random numbers between 0 and 10

# Create another variable (Y) that has a linear relationship with X, plus some noise
Y = 2 * X + 5 + np.random.randn(100) * 2 # Y = 2X + 5 + noise

df_relationship = pd.DataFrame({
    'Variabel_X': X,
    'Variabel_Y': Y
})

display(df_relationship.head())
```

### Visualisasi Hubungan Antar Variabel: Scatter Plot

Scatter plot (diagram pencar) adalah jenis grafik yang menggunakan titik-titik (dots) untuk merepresentasikan nilai-nilai dari dua variabel numerik yang berbeda. Setiap titik pada scatter plot mewakili satu observasi, dengan posisi horizontal (sumbu X) menunjukkan nilai dari satu variabel, dan posisi vertikal (sumbu Y) menunjukkan nilai dari variabel lainnya.

Scatter plot sangat efektif untuk:
*   Mengidentifikasi hubungan atau korelasi antara dua variabel (misalnya, positif, negatif, tidak ada).
*   Melihat pola distribusi data.
*   Mendeteksi adanya outlier atau anomali dalam data.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Variabel_X', y='Variabel_Y', data=df_relationship)
plt.title('Scatter Plot Hubungan Antara Variabel X dan Variabel Y')
plt.xlabel('Variabel X')
plt.ylabel('Variabel Y')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

#### Cara Membaca Scatter Plot

Untuk membaca scatter plot dan memahami hubungan antar variabel:

1.  **Arah Hubungan (Korelasi):**
    *   **Positif:** Jika titik-titik cenderung membentuk pola naik dari kiri bawah ke kanan atas, menunjukkan bahwa ketika nilai Variabel X meningkat, nilai Variabel Y juga cenderung meningkat.
    *   **Negatif:** Jika titik-titik cenderung membentuk pola turun dari kiri atas ke kanan bawah, menunjukkan bahwa ketika nilai Variabel X meningkat, nilai Variabel Y cenderung menurun.
    *   **Tidak Ada Hubungan:** Jika titik-titik tersebar secara acak tanpa pola yang jelas, menunjukkan tidak ada hubungan linier yang signifikan antara kedua variabel.

2.  **Kekuatan Hubungan:**
    *   **Kuat:** Jika titik-titik sangat berdekatan satu sama lain dan membentuk pola yang jelas (mendekati garis lurus).
    *   **Sedang:** Jika titik-titik memiliki pola yang terlihat tetapi sedikit lebih tersebar.
    *   **Lemah:** Jika pola sulit diidentifikasi dan titik-titik sangat tersebar.

3.  **Bentuk Hubungan:**
    *   **Linier:** Jika titik-titik cenderung mengikuti pola garis lurus.
    *   **Non-linier:** Jika titik-titik membentuk pola melengkung (misalnya, kurva).

4.  **Pencilan (Outlier):** Titik-titik yang terletak jauh dari pola umum sebagian besar data bisa menjadi outlier. Ini bisa menunjukkan kesalahan data atau kejadian yang tidak biasa dan perlu diselidiki lebih lanjut.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Generate dummy data for time series
np.random.seed(42) # for reproducibility

# Create a date range for 1 year (365 days)
dates = pd.date_range(start='2023-01-01', periods=365, freq='D')

# Create a numerical variable with an increasing trend and some random fluctuations
trend = np.linspace(50, 150, len(dates)) # Increasing trend from 50 to 150
noise = np.random.normal(loc=0, scale=10, size=len(dates))

# Combine trend and noise
value = trend + noise

df_timeseries = pd.DataFrame({
    'Tanggal': dates,
    'Nilai': value
})

display(df_timeseries.head())
```

### Visualisasi Tren Waktu: Grafik Garis (Line Plot)

Grafik garis (line plot) adalah jenis visualisasi yang paling umum digunakan untuk menampilkan data deret waktu (time series). Ini sangat efektif untuk menunjukkan bagaimana suatu variabel berubah atau berkembang seiring waktu. Setiap titik pada garis mewakili nilai variabel pada waktu tertentu, dan garis yang menghubungkan titik-titik tersebut menunjukkan tren atau pola perubahan.

Grafik garis sangat berguna untuk:
*   Mengidentifikasi tren jangka panjang (meningkat, menurun, stabil).
*   Melihat pola musiman atau siklus.
*   Mendeteksi anomali atau perubahan mendadak.
*   Membandingkan kinerja dari waktu ke waktu.

```python
plt.figure(figsize=(14, 7))
sns.lineplot(x='Tanggal', y='Nilai', data=df_timeseries)
plt.title('Grafik Tren Waktu Nilai Seiring Berjalannya Waktu')
plt.xlabel('Tanggal')
plt.ylabel('Nilai')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45) # Rotate x-axis labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

#### Cara Membaca Grafik Garis (Tren Waktu)

Untuk membaca grafik garis dan memahami tren waktu:

1.  **Sumbu X (Horizontal):** Selalu merepresentasikan waktu (tanggal, bulan, tahun, jam, dll.), berjalan dari kiri ke kanan.
2.  **Sumbu Y (Vertikal):** Merepresentasikan nilai dari variabel yang sedang diukur atau diamati (misalnya, 'Nilai' dalam contoh ini).
3.  **Arah Garis:**
    *   **Tren Naik:** Jika garis secara umum bergerak ke atas dari kiri ke kanan, menunjukkan nilai variabel meningkat seiring waktu.
    *   **Tren Turun:** Jika garis secara umum bergerak ke bawah dari kiri ke kanan, menunjukkan nilai variabel menurun seiring waktu.
    *   **Stabil/Datar:** Jika garis relatif lurus atau berfluktuasi di sekitar level tertentu, menunjukkan nilai variabel relatif stabil.
4.  **Pola dan Fluktuasi:**
    *   **Musiman:** Cari pola yang berulang pada interval waktu yang teratur (misalnya, setiap minggu, setiap bulan, setiap tahun). Ini menunjukkan adanya pengaruh musiman.
    *   **Siklus:** Mirip dengan musiman tetapi dengan durasi yang tidak teratur dan seringkali lebih panjang (misalnya, siklus ekonomi).
    *   **Ketidakteraturan/Kebisingan:** Fluktuasi acak yang tidak mengikuti pola tertentu. Ini adalah 'noise' dalam data.
5.  **Titik Anomali (Outlier):** Perhatikan titik-titik yang menyimpang jauh dari pola umum garis. Ini bisa menjadi kejadian penting, kesalahan data, atau outlier yang perlu diselidiki.

### Visualisasi Komposisi

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Generate dummy data for compositional visualization
np.random.seed(42) # for reproducibility

# Main categories
main_categories = ['Wilayah A', 'Wilayah B', 'Wilayah C']

# Sub-categories for each main category
sub_categories_A = ['Produk X', 'Produk Y', 'Produk Z']
sub_categories_B = ['Produk P', 'Produk Q']
sub_categories_C = ['Produk R', 'Produk S', 'Produk T', 'Produk U']

data = []

for main_cat in main_categories:
    if main_cat == 'Wilayah A':
        sub_cats = sub_categories_A
    elif main_cat == 'Wilayah B':
        sub_cats = sub_categories_B
    else:
        sub_cats = sub_categories_C

    # Generate values for sub-categories that sum up to a total for the main category
    # For simplicity, let's just assign random values and then we can calculate proportions
    for sub_cat in sub_cats:
        data.append({
            'Kategori_Utama': main_cat,
            'Sub_Kategori': sub_cat,
            'Penjualan': np.random.randint(100, 500) # Random sales value
        })

df_composition = pd.DataFrame(data)

display(df_composition.head())
```

### 1. Stacked Bar Chart (Grafik Batang Bertumpuk)

Stacked bar chart adalah jenis grafik batang yang membagi setiap batang menjadi beberapa segmen. Setiap segmen dalam batang mewakili bagian dari total, dan tinggi atau panjang total batang menunjukkan nilai total untuk kategori tersebut. Grafik ini sangat berguna untuk memvisualisasikan komposisi kategori utama berdasarkan sub-kategori, serta membandingkan total antar kategori utama.

```python
# Calculate total sales per Main Category for overall bar height
df_total_sales = df_composition.groupby('Kategori_Utama')['Penjualan'].sum().reset_index()

plt.figure(figsize=(12, 7))
sns.barplot(
    x='Kategori_Utama',
    y='Penjualan',
    hue='Sub_Kategori',
    data=df_composition,
    palette='viridis',
    dodge=False # Important for stacked bars
)
plt.title('Stacked Bar Chart Penjualan per Wilayah dan Produk')
plt.xlabel('Kategori Utama (Wilayah)')
plt.ylabel('Total Penjualan')
plt.grid(axis='y', alpha=0.75)
plt.legend(title='Sub Kategori (Produk)')
plt.tight_layout()
plt.show()
```

#### Cara Membaca Stacked Bar Chart

Untuk membaca stacked bar chart:

1.  **Setiap Batang:** Mewakili satu kategori utama (misalnya, 'Wilayah A', 'Wilayah B'). Tinggi total batang menunjukkan jumlah total atau nilai agregat untuk kategori utama tersebut (misalnya, total penjualan wilayah tersebut).
2.  **Segmen dalam Batang:** Setiap warna atau segmen dalam satu batang mewakili proporsi atau kontribusi dari sub-kategori (misalnya, 'Produk X', 'Produk Y') terhadap total kategori utama.
3.  **Sumbu X (Horizontal):** Menampilkan kategori utama yang sedang dibandingkan.
4.  **Sumbu Y (Vertikal):** Menunjukkan nilai kuantitatif yang sedang diukur (misalnya, 'Total Penjualan').
5.  **Perbandingan:**
    *   **Antar Kategori Utama:** Bandingkan tinggi total setiap batang untuk melihat kategori utama mana yang memiliki nilai agregat tertinggi atau terendah.
    *   **Komposisi Internal:** Lihat proporsi setiap segmen dalam satu batang untuk memahami bagaimana sub-kategori berkontribusi terhadap total kategori utama. Misalnya, produk mana yang paling dominan di 'Wilayah A'.
    *   **Perbandingan Sub-kategori antar Kategori Utama:** Bandingkan segmen dengan warna yang sama di berbagai batang untuk melihat bagaimana kontribusi sub-kategori tertentu bervariasi antar kategori utama (meskipun ini bisa lebih sulit jika ada banyak segmen).

### 2. Pie Chart (Diagram Lingkaran)

Pie chart adalah grafik melingkar yang dibagi menjadi 'irisan' atau segmen untuk mengilustrasikan proporsi numerik. Dalam pie chart, setiap irisan mewakili proporsi kategori tertentu dari keseluruhan. Jumlah dari semua irisan harus sama dengan 100%. Grafik ini paling efektif untuk memvisualisasikan komposisi atau pangsa dari beberapa kategori dalam satu keseluruhan tunggal, terutama ketika jumlah kategori tidak terlalu banyak.

```python
# Calculate total sales for each Main Category to be used in the pie chart
df_pie_data = df_composition.groupby('Kategori_Utama')['Penjualan'].sum().reset_index()

plt.figure(figsize=(8, 8))
plt.pie(
    df_pie_data['Penjualan'],
    labels=df_pie_data['Kategori_Utama'],
    autopct='%1.1f%%', # Format for displaying percentage
    startangle=90,
    colors=sns.color_palette('pastel')
)
plt.title('Komposisi Penjualan Berdasarkan Kategori Utama')
plt.axis('equal') # Ensures the pie chart is circular
plt.show()
```

#### Cara Membaca Pie Chart

Untuk membaca pie chart:

1.  **Lingkaran Keseluruhan:** Mewakili total keseluruhan (100%) dari apa yang sedang dianalisis (misalnya, total penjualan dari semua kategori utama).
2.  **Irisan (Slices):** Setiap irisan mewakili satu kategori. Ukuran irisan (luas dan sudutnya) secara langsung proporsional dengan proporsi kategori tersebut terhadap total keseluruhan.
3.  **Label dan Persentase:** Umumnya, setiap irisan akan memiliki label yang menunjukkan nama kategori dan persentasenya dari total. Persentase ini adalah cara paling langsung untuk memahami kontribusi relatif setiap kategori.
4.  **Perbandingan:** Bandingkan ukuran irisan yang berbeda. Irisan yang lebih besar menunjukkan kontribusi atau pangsa yang lebih besar terhadap keseluruhan. Ini memudahkan identifikasi kategori mana yang paling dominan atau paling kecil.

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from collections import Counter
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

def visualisasi_teks_lengkap(corpus, label=None, top_n=30):
    """
    Panel visualisasi komprehensif untuk analisis teks.
    corpus: list of preprocessed text strings
    label: optional labels for comparison
    """
    fig = plt.figure(figsize=(16, 10))
    fig.suptitle('Analisis Visual Korpus Teks',
                 fontweight='bold', color='#1A3C6E', fontsize=14)

    # ── Panel 1: Word Cloud ──
    ax1 = fig.add_subplot(2, 3, 1)
    teks_gabung = ' '.join(corpus)
    wc = WordCloud(
        width=600, height=400,
        background_color='white',
        colormap='Blues',
        max_words=100,
        prefer_horizontal=0.9,
        collocations=False,
    ).generate(teks_gabung)
    ax1.imshow(wc, interpolation='bilinear')
    ax1.axis('off')
    ax1.set_title('Word Cloud')

    # ── Panel 2: Frekuensi kata top-N ──
    ax2 = fig.add_subplot(2, 3, 2)
    semua_kata = ' '.join(corpus).split()
    freq = Counter(semua_kata).most_common(top_n)
    kata, hitung = zip(*freq)
    y_pos = range(len(kata))
    ax2.barh(y_pos, hitung, color='#1A3C6E', alpha=0.8)
    ax2.set_yticks(y_pos)
    ax2.set_yticklabels(kata, fontsize=8)
    ax2.set_xlabel('Frekuensi')
    ax2.set_title(f'Top {top_n} Kata')
    ax2.invert_yaxis()
    ax2.spines[['top','right']].set_visible(False)

    # ── Panel 3: Distribusi panjang dokumen ──
    ax3 = fig.add_subplot(2, 3, 3)
    panjang = [len(t.split()) for t in corpus]
    ax3.hist(panjang, bins=30, color='#1A3C6E', alpha=0.7, edgecolor='white')
    ax3.axvline(np.mean(panjang), color='#D95F20', ls='--', lw=2,
                label=f'Rata-rata = {np.mean(panjang):.0f}')
    ax3.axvline(np.median(panjang), color='#2E7D32', ls='--', lw=2,
                label=f'Median = {np.median(panjang):.0f}')
    ax3.set_xlabel('Jumlah Kata per Dokumen')
    ax3.set_ylabel('Frekuensi')
    ax3.set_title('Distribusi Panjang Dokumen')
    ax3.legend(fontsize=8)
    ax3.spines[['top','right']].set_visible(False)

    # ── Panel 4–6: Perbandingan TF-IDF per label ──
    if label is not None:
        kelas = sorted(set(label))
        tfidf = TfidfVectorizer(max_features=15, ngram_range=(1,2))
        for i, kls in enumerate(kelas[:3]):
            ax = fig.add_subplot(2, 3, 4+i)
            teks_kls = [t for t, l in zip(corpus, label) if l == kls]
            tfidf.fit(teks_kls)
            scores = zip(tfidf.get_feature_names_out(),
                         tfidf.idf_)
            scores = sorted(scores, key=lambda x: -x[1])[:15]
            w, s = zip(*scores)
            ax.barh(range(len(w)), s, color='#D95F20', alpha=0.8)
            ax.set_yticks(range(len(w)))
            ax.set_yticklabels(w, fontsize=8)
            ax.set_title(f'IDF Tertinggi: {kls}', fontsize=10)
            ax.invert_yaxis()
            ax.spines[['top','right']].set_visible(False)

    plt.tight_layout()
    # plt.savefig('output/visualisasi_teks.pdf', bbox_inches='tight') # Removed to prevent FileNotFoundError
    plt.show()
```

Puncak yang tinggi dan sempit menunjukkan dokumen memiliki panjang yang relatif seragam, sementara puncak yang lebar atau banyak menunjukkan variasi yang signifikan.

##### 4. TF-IDF Tertinggi per Label (jika ada label)

*   **Teori:** TF-IDF (Term Frequency-Inverse Document Frequency) adalah metrik numerik yang dimaksudkan untuk mencerminkan seberapa penting sebuah kata bagi sebuah dokumen dalam sebuah korpus. Kata-kata dengan TF-IDF tinggi adalah kata-kata yang sering muncul dalam dokumen tertentu tetapi jarang di seluruh korpus, sehingga menunjukkan relevansi yang tinggi untuk dokumen tersebut. Jika label disediakan, panel ini membandingkan kata-kata dengan nilai TF-IDF tertinggi untuk setiap label, membantu mengidentifikasi kata kunci yang membedakan antar kategori.
*   **Cara Membaca:** Setiap grafik batang mewakili satu kategori label. Panjang batang menunjukkan nilai TF-IDF dari kata kunci yang terkait. Kata-kata dengan batang yang lebih panjang adalah yang paling unik dan membedakan untuk kategori tersebut dibandingkan dengan kategori lain. Ini sangat berguna untuk analisis topik atau klasifikasi teks, karena menunjukkan istilah mana yang merupakan indikator kuat dari suatu kategori.

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import (
    confusion_matrix, ConfusionMatrixDisplay,
    roc_curve, auc, precision_recall_curve
)
from sklearn.inspection import permutation_importance
import pandas as pd

def dashboard_evaluasi_model(model, X_test, y_test, kelas,
                              kolom_fitur, nama_model='Model'):
    """
    Dashboard evaluasi komprehensif untuk model klasifikasi.
    Menampilkan: confusion matrix, ROC curves, PR curve, feature importance.
    """
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test) if hasattr(model, 'predict_proba') else None

    fig = plt.figure(figsize=(16, 10))
    fig.suptitle(f'Dashboard Evaluasi — {nama_model}',
                 fontweight='bold', color='#1A3C6E', fontsize=14)

    gs = plt.GridSpec(2, 3, figure=fig, hspace=0.4, wspace=0.35)

    # ── Panel 1: Confusion Matrix ──
    ax1 = fig.add_subplot(gs[0, 0])
    cm = confusion_matrix(y_test, y_pred, normalize='true')
    disp = ConfusionMatrixDisplay(cm, display_labels=kelas)
    disp.plot(ax=ax1, cmap='Blues', colorbar=False, values_format='.2f')
    ax1.set_title('Confusion Matrix\n(Proporsi per Kelas Aktual)')
    ax1.set_xlabel('Kelas Prediksi')
    ax1.set_ylabel('Kelas Aktual')

    # ── Panel 2: ROC Curves (per kelas) ──
    ax2 = fig.add_subplot(gs[0, 1])
    if y_prob is not None:
        from sklearn.preprocessing import label_binarize
        y_bin = label_binarize(y_test, classes=range(len(kelas)))
        warna_roc = ['#1A3C6E','#D95F20','#2E7D32','#B71C1C']
        for i, (kls, w) in enumerate(zip(kelas, warna_roc)):
            fpr, tpr, _ = roc_curve(y_bin[:, i], y_prob[:, i])
            auc_kls     = auc(fpr, tpr)
            ax2.plot(fpr, tpr, color=w, lw=1.5,
                     label=f'{kls} (AUC={auc_kls:.2f})')
        ax2.plot([0,1],[0,1],'k--',lw=0.8,alpha=0.5)
        ax2.set_xlabel('False Positive Rate')
        ax2.set_ylabel('True Positive Rate')
        ax2.set_title('ROC Curves per Kelas')
        ax2.legend(fontsize=8, loc='lower right')
        ax2.spines[['top','right']].set_visible(False)

    # ── Panel 3: Distribusi prediksi vs aktual ──
    ax3 = fig.add_subplot(gs[0, 2])
    df_pred = pd.DataFrame({'aktual': y_test, 'prediksi': y_pred})
    pct_benar = (df_pred.groupby('aktual')
                 .apply(lambda g: (g['aktual']==g['prediksi']).mean())
                 * 100).reset_index(name='pct_benar')
    ax3.barh(pct_benar['aktual'].astype(str),
             pct_benar['pct_benar'],
             color=['#2E7D32' if p >= 70 else '#B71C1C'
                    for p in pct_benar['pct_benar']],
             alpha=0.8)
    ax3.axvline(70, color='gray', ls='--', lw=1)
    ax3.set_xlabel('% Prediksi Benar')
    ax3.set_title('Akurasi per Kelas')
    ax3.set_xlim(0, 105)
    ax3.spines[['top','right']].set_visible(False)

    # ── Panel 4: Feature Importance (Permutation) ──
    ax4 = fig.add_subplot(gs[1, :2])
    from sklearn.metrics import f1_score
    perm = permutation_importance(
        model, X_test, y_test,
        scoring   = 'f1_macro',
        n_repeats = 15,
        random_state = 42,
        n_jobs    = -1,
    )
    fi_df = pd.DataFrame({
        'fitur'    : kolom_fitur,
        'importance': perm.importances_mean,
        'std'       : perm.importances_std,
    }).sort_values('importance', ascending=True).tail(15)

    ax4.barh(fi_df['fitur'], fi_df['importance'],
             xerr=fi_df['std'], color='#1A3C6E', alpha=0.8,
             ecolor='#CBD8E8', capsize=3)
    ax4.axvline(0, color='gray', lw=0.8)
    ax4.set_xlabel('Penurunan F1-macro (Permutation Importance)')
    ax4.set_title('Top-15 Feature Importance (Permutation)')
    ax4.spines[['top','right']].set_visible(False)

    # ── Panel 5: Distribusi probabilitas prediksi ──
    ax5 = fig.add_subplot(gs[1, 2])
    if y_prob is not None:
        conf_benar = y_prob.max(axis=1)[y_pred == y_test]
        conf_salah = y_prob.max(axis=1)[y_pred != y_test]
        ax5.hist(conf_benar, bins=20, alpha=0.7, color='#2E7D32',
                 label='Prediksi Benar', density=True)
        ax5.hist(conf_salah, bins=20, alpha=0.7, color='#B71C1C',
                 label='Prediksi Salah', density=True)
        ax5.set_xlabel('Probabilitas Prediksi Tertinggi')
        ax5.set_ylabel('Densitas')
        ax5.set_title('Kepercayaan Model')
        ax5.legend(fontsize=8)
        ax5.spines[['top','right']].set_visible(False)

    # plt.savefig(f'output/dashboard_{nama_model.replace(" ","_")}.pdf',
    #             bbox_inches='tight') # Removed to prevent FileNotFoundError
    plt.show()
```

### Dashboard Evaluasi Model Klasifikasi

Fungsi `dashboard_evaluasi_model` dirancang untuk memberikan tinjauan komprehensif tentang kinerja model klasifikasi. Ini mencakup metrik dan visualisasi penting seperti Confusion Matrix, Kurva ROC, Akurasi per Kelas, Pentingnya Fitur (Feature Importance), dan Kepercayaan Model.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification

# Generate dummy classification data
X, y = make_classification(
    n_samples=1000,
    n_features=10,
    n_informative=5,
    n_redundant=2,
    n_classes=3, # Multi-class classification
    random_state=42
)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Define feature column names
kolom_fitur = [f'Fitur_{i+1}' for i in range(X.shape[1])]

# Define class labels
kelas = ['Kelas A', 'Kelas B', 'Kelas C']

# Create a dummy classification model (Logistic Regression)
model_dummy = LogisticRegression(random_state=42, multi_class='ovr', solver='liblinear')
model_dummy.fit(X_train, y_train)

# Call the dashboard evaluation function
dashboard_evaluasi_model(
    model=model_dummy,
    X_test=X_test,
    y_test=y_test,
    kelas=kelas,
    kolom_fitur=kolom_fitur,
    nama_model='Model Klasifikasi Dummy'
)
```

##### 1. Confusion Matrix (Matriks Kebingungan)

*   **Teori:** Confusion Matrix adalah tabel yang digunakan untuk menggambarkan kinerja model klasifikasi pada sekumpulan data uji yang nilai benarnya diketahui. Ini memungkinkan visualisasi kinerja algoritma.
    *   `True Positive (TP)`: Diprediksi positif, dan memang positif.
    *   `True Negative (TN)`: Diprediksi negatif, dan memang negatif.
    *   `False Positive (FP)`: Diprediksi positif, tapi sebenarnya negatif (Type I error).
    *   `False Negative (FN)`: Diprediksi negatif, tapi sebenarnya positif (Type II error).
    *   Visualisasi ini biasanya menampilkan *proporsi* per kelas aktual, yang berarti setiap baris akan berjumlah 1.
*   **Cara Membaca:**
    *   **Diagonal:** Angka pada diagonal (dari kiri atas ke kanan bawah) menunjukkan proporsi prediksi yang benar untuk setiap kelas. Semakin tinggi angkanya (mendekati 1.00), semakin baik model dalam mengklasifikasikan kelas tersebut.
    *   **Di Luar Diagonal:** Angka di luar diagonal menunjukkan kesalahan klasifikasi. Misalnya, nilai di baris 'Kelas A', kolom 'Kelas B' menunjukkan proporsi instans 'Kelas A' yang salah diprediksi sebagai 'Kelas B'.
    *   **Perhatikan Baris:** Setiap baris mewakili kelas aktual. Jumlah proporsi di setiap baris harus 1.00.

##### 2. ROC Curves per Kelas (Kurva Karakteristik Operasi Penerima)

*   **Teori:** Kurva ROC adalah plot yang menggambarkan kinerja model klasifikasi pada semua ambang klasifikasi. Kurva ini memplot True Positive Rate (TPR atau Sensitivitas) terhadap False Positive Rate (FPR atau 1-Spesifisitas) pada berbagai pengaturan ambang batas. Area Under the Curve (AUC) adalah ukuran kinerja agregat dari semua kemungkinan ambang batas klasifikasi. Nilai AUC berkisar dari 0 hingga 1; model yang sepenuhnya acak memiliki AUC 0.5, sedangkan model yang sempurna memiliki AUC 1.0.
*   **Cara Membaca:**
    *   **Bentuk Kurva:** Semakin dekat kurva ke sudut kiri atas grafik, semakin baik kinerja model. Ini menunjukkan tingkat TPR yang tinggi dengan FPR yang rendah.
    *   **Nilai AUC:** Perhatikan nilai AUC untuk setiap kelas (ditampilkan dalam legenda). Semakin tinggi nilai AUC, semakin baik model dalam membedakan antara kelas positif dan negatif untuk kelas tersebut. AUC 0.90 atau lebih tinggi umumnya dianggap sangat baik.

##### 3. Akurasi per Kelas (Persentase Prediksi Benar)

*   **Teori:** Grafik batang ini menunjukkan persentase instans yang diprediksi dengan benar untuk setiap kelas aktual. Ini adalah metrik yang lebih mudah diinterpretasikan daripada confusion matrix untuk melihat seberapa akurat model pada masing-masing kategori target.
*   **Cara Membaca:**
    *   **Tinggi Batang:** Setiap batang mewakili satu kelas, dan tingginya menunjukkan persentase prediksi yang benar untuk kelas tersebut. Batang yang lebih tinggi menandakan akurasi yang lebih baik untuk kelas tersebut.
    *   **Garis Referensi:** Garis putus-putus pada 70% dapat menjadi patokan cepat untuk mengidentifikasi kelas mana yang memiliki kinerja yang kurang memuaskan (di bawah 70%). Warna batang juga dapat menunjukkan ini (misalnya, hijau untuk >= 70%, merah untuk < 70%).

##### 4. Top-15 Feature Importance (Permutation)

*   **Teori:** Permutation Importance adalah teknik untuk mengukur seberapa penting suatu fitur untuk model terlatih. Ini dilakukan dengan mengocok (permutasi) nilai-nilai suatu fitur dan mengamati seberapa besar penurunan skor model. Fitur yang penting akan menyebabkan penurunan skor yang signifikan ketika diacak. Ini adalah metode yang 'model-agnostic' dan dapat memberikan wawasan yang lebih andal tentang pentingnya fitur dibandingkan metode berbasis koefisien.
*   **Cara Membaca:**
    *   **Panjang Batang:** Semakin panjang batang untuk sebuah fitur, semakin besar penurunan F1-macro (atau metrik skor lain yang digunakan) ketika fitur tersebut diacak, yang berarti fitur tersebut lebih penting bagi model.
    *   **Error Bars:** Garis horizontal tipis (error bars) menunjukkan standar deviasi dari penurunan skor di beberapa pengulangan permutasi. Error bars yang pendek menunjukkan bahwa pentingnya fitur tersebut konsisten.
    *   **Sumbu X:** Menunjukkan 'Penurunan F1-macro', yang merupakan ukuran kontribusi fitur terhadap kinerja model.

##### 5. Kepercayaan Model (Distribusi Probabilitas Prediksi)

*   **Teori:** Histogram ini membandingkan distribusi probabilitas prediksi tertinggi yang diberikan oleh model untuk kasus di mana model memprediksi dengan benar dan kasus di mana model memprediksi salah. Ini memberikan wawasan tentang seberapa 'percaya diri' model dalam prediksinya dan apakah model cenderung overconfident atau underconfident.
*   **Cara Membaca:**
    *   **Prediksi Benar (Hijau):** Melihat distribusi ini, Anda ingin melihat puncak yang tinggi mendekati 1.0, menunjukkan bahwa model sangat percaya diri ketika membuat prediksi yang benar.
    *   **Prediksi Salah (Merah):** Untuk prediksi yang salah, Anda ingin melihat distribusi yang puncaknya lebih rendah atau lebih tersebar, idealnya menjauh dari 1.0. Jika ada banyak prediksi salah dengan probabilitas tinggi (mendekati 1.0), ini menunjukkan model terlalu percaya diri pada prediksinya yang salah.
    *   **Perbandingan:** Bandingkan kedua distribusi. Idealnya, distribusi untuk prediksi benar harus lebih terkonsentrasi di probabilitas tinggi dibandingkan dengan distribusi untuk prediksi salah.

## Tutorial Matplotlib dan Seaborn dengan Dataset Transaksi

Bagian ini akan menunjukkan beberapa contoh visualisasi data menggunakan library Matplotlib dan Seaborn, memanfaatkan dataset `df` yang telah kita siapkan sebelumnya. Kita akan membuat ulang `df` di sini untuk memastikan setiap contoh dapat berjalan secara mandiri.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Memastikan direktori 'data' ada dan file CSV tersedia
if not os.path.exists('data'):
    os.makedirs('data')
    print("Direktori 'data' berhasil dibuat.")

# Membuat file CSV dummy jika belum ada
if not os.path.exists('data/data_transaksi.csv'):
    np.random.seed(42)
    dummy_dates = (pd.date_range(start='2023-01-01', periods=10, freq='D').tolist() +
                  ['2023-13-01', 'invalid-date', '2023-02-15'])
    dummy_harga = np.random.randint(10000, 100000, size=13)
    dummy_jumlah = np.random.randint(1, 10, size=13)
    df_dummy_transaksi = pd.DataFrame({
        'tanggal': dummy_dates,
        'harga': dummy_harga,
        'jumlah': dummy_jumlah
    })
    df_dummy_transaksi.to_csv('data/data_transaksi.csv', index=False)
    print("File 'data/data_transaksi.csv' berhasil dibuat dengan data dummy.")
else:
    print("File 'data/data_transaksi.csv' sudah ada.")

# Memuat ulang DataFrame df dan menambahkan kolom yang diperlukan
df = pd.read_csv("data/data_transaksi.csv")
df["tanggal"] = pd.to_datetime(df["tanggal"], errors="coerce")
df["total_transaksi"] = df["harga"] * df["jumlah"]

# Menambahkan kolom 'kategori' dummy untuk demonstrasi
np.random.seed(42) # untuk reproduktifitas
df['kategori'] = np.random.choice(['Elektronik', 'Pakaian', 'Makanan', 'Otomotif'], size=len(df))

print("DataFrame 'df' berhasil disiapkan dengan kolom 'total_transaksi' dan 'kategori'.")
display(df.head())
```

### 1. Seaborn Boxplot: Sebaran Total Transaksi per Kategori

Boxplot adalah visualisasi yang sangat berguna untuk memahami distribusi data numerik dalam setiap kelompok kategori. Ini menunjukkan median, kuartil (Q1 dan Q3), serta potensi outlier.

Dalam contoh ini, kita akan menggunakan Seaborn untuk melihat bagaimana 'total_transaksi' tersebar di berbagai 'kategori' produk.

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='kategori', y='total_transaksi', data=df, palette='pastel')
plt.title('Sebaran Total Transaksi per Kategori (Seaborn Boxplot)')
plt.xlabel('Kategori Produk')
plt.ylabel('Total Transaksi')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

#### Cara Membaca Seaborn Boxplot

1.  **Kotak (Box):** Bagian tengah kotak menunjukkan **Interquartile Range (IQR)**, yaitu 50% data tengah. Garis di dalam kotak adalah **median** (nilai tengah). Tepi bawah kotak adalah **kuartil pertama (Q1)**, dan tepi atas adalah **kuartil ketiga (Q3)**.
2.  **Garis Kumis (Whiskers):** Garis yang memanjang dari kotak. Secara default, kumis ini biasanya mencapai nilai data terjauh dalam batas 1.5 kali IQR dari Q1 dan Q3. Ini menunjukkan rentang data `normal`.
3.  **Titik-titik (Outliers):** Titik-titik individual di luar kumis adalah **outlier**, yaitu nilai-nilai yang jauh dari sebagian besar data dan mungkin memerlukan penyelidikan lebih lanjut.

Dari boxplot di atas, Anda dapat membandingkan:
*   **Median:** Kategori mana yang memiliki median `total_transaksi` lebih tinggi atau lebih rendah.
*   **Sebaran (Spread):** Seberapa lebar kotak (IQR) dan panjang kumis menunjukkan variabilitas `total_transaksi` dalam setiap kategori.
*   **Outlier:** Adanya titik-titik di luar kumis menunjukkan transaksi yang sangat besar atau sangat kecil untuk kategori tersebut.

### 2. Matplotlib Line Plot: Tren Harian Total Transaksi

Line plot (grafik garis) sangat cocok untuk memvisualisasikan data deret waktu (time series) dan menunjukkan tren atau pola perubahan suatu variabel seiring waktu. Untuk ini, kita akan melihat tren `total_transaksi` harian.

```python
# Agregasi total transaksi harian
df_daily_transactions = df.groupby('tanggal')['total_transaksi'].sum().reset_index()

plt.figure(figsize=(12, 6))
plt.plot(df_daily_transactions['tanggal'], df_daily_transactions['total_transaksi'], marker='o', linestyle='-', color='skyblue', markersize=4)
plt.title('Tren Harian Total Transaksi (Matplotlib Line Plot)')
plt.xlabel('Tanggal')
plt.ylabel('Total Transaksi Harian')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Cara Membaca Matplotlib Line Plot

1.  **Sumbu X (Horizontal):** Mewakili waktu (dalam kasus ini, 'tanggal'). Data bergerak maju seiring Anda bergerak ke kanan.
2.  **Sumbu Y (Vertikal):** Mewakili nilai variabel yang diukur (dalam kasus ini, 'Total Transaksi Harian').
3.  **Arah Garis:** Arah garis menghubungkan titik-titik data secara berurutan waktu. Ini menunjukkan tren: jika garis naik, nilai variabel meningkat; jika garis turun, nilai variabel menurun.
4.  **Pola dan Fluktuasi:** Anda dapat mengamati pola musiman, siklus, atau fluktuasi acak dalam data. Misalnya, apakah ada puncak atau lembah yang teratur, atau apakah ada lonjakan mendadak.

Dari line plot di atas, Anda dapat mengidentifikasi:
*   **Tren Keseluruhan:** Apakah total transaksi harian cenderung meningkat, menurun, atau tetap stabil selama periode waktu yang diamati.
*   **Puncak dan Lembah:** Kapan terjadi transaksi harian tertinggi atau terendah.
*   **Anomali:** Setiap titik data yang menyimpang jauh dari tren umum mungkin mengindikasikan peristiwa atau anomali khusus.

### 1. Plotly Interactive Line Chart: Tren Harian Total Transaksi

Plotly adalah *library* visualisasi yang powerful untuk membuat grafik interaktif yang dapat di-zoom, di-pan, dan menampilkan *tooltip* informasi. Kita akan menggunakannya untuk melihat tren `total_transaksi` harian dari dataset `df`.

```python
import plotly.express as px

# Agregasi total transaksi harian dari df
df_daily_transactions_plotly = df.groupby('tanggal')['total_transaksi'].sum().reset_index()

# Membuat interactive line chart menggunakan Plotly Express
fig = px.line(
    df_daily_transactions_plotly,
    x='tanggal',
    y='total_transaksi',
    title='Tren Harian Total Transaksi (Plotly Interactive Line Chart)',
    labels={'tanggal': 'Tanggal', 'total_transaksi': 'Total Transaksi Harian'}
)

# Menambahkan custom layout untuk tampilan yang lebih baik
fig.update_layout(
    xaxis_title='Tanggal',
    yaxis_title='Total Transaksi Harian',
    hovermode='x unified'
)

fig.show()
```

#### Cara Membaca Plotly Interactive Line Chart

Visualisasi interaktif Plotly memiliki keunggulan dibandingkan grafik statis. Anda dapat:

1.  **Hover (Arahkan Kursor):** Arahkan kursor ke titik mana pun pada garis untuk melihat detail nilai `tanggal` dan `total_transaksi` spesifik pada titik tersebut. Ini sangat membantu untuk mendapatkan informasi tepat tanpa perlu membaca label sumbu secara manual.
2.  **Zoom dan Pan:** Gunakan tombol di pojok kanan atas grafik (atau geser area pada grafik) untuk memperbesar (zoom) bagian tertentu dari grafik. Anda juga bisa menggeser (pan) grafik untuk menjelajahi area yang berbeda.
3.  **Reset Axes:** Klik tombol 'Reset axes' (gambar rumah) untuk mengembalikan grafik ke tampilan asli setelah melakukan zoom atau pan.
4.  **Toggle Data:** Jika ada beberapa garis (misalnya, untuk beberapa kategori), Anda bisa mengklik legenda untuk menyembunyikan atau menampilkan garis tertentu.

Dari grafik ini, Anda dapat mengamati tren, puncak, lembah, dan anomali dalam data total transaksi harian secara lebih dinamis.

### 2. Altair Interactive Scatter Plot: Hubungan Total Transaksi dan Jumlah

Altair adalah *library* deklaratif untuk membuat visualisasi statistik yang interaktif. Ini sangat baik untuk eksplorasi data karena memungkinkan pemilihan dan *brushing* data yang mudah. Kita akan membuat *scatter plot* untuk melihat hubungan antara `total_transaksi` dan `jumlah` dari dataset `df`, dengan warna berdasarkan `kategori`.

```python
import altair as alt

# Membuat interactive scatter plot menggunakan Altair
alt.Chart(df).mark_point().encode(
    x='jumlah:Q', # 'Q' denotes quantitative data
    y='total_transaksi:Q',
    color='kategori:N', # 'N' denotes nominal data (categorical)
    tooltip=['tanggal', 'harga', 'jumlah', 'total_transaksi', 'kategori'] # Informasi saat hover
).properties(
    title='Hubungan Jumlah dan Total Transaksi per Kategori (Altair Scatter Plot)'
).interactive()
```

#### Cara Membaca Altair Interactive Scatter Plot

Altair memungkinkan Anda untuk berinteraksi langsung dengan visualisasi, seperti:

1.  **Hover (Arahkan Kursor):** Arahkan kursor ke setiap titik untuk melihat detail data dari transaksi spesifik (tanggal, harga, jumlah, total_transaksi, kategori).
2.  **Zoom dan Pan:** Gunakan gulir mouse (scroll) untuk *zoom in* dan *zoom out*, serta klik-dan-seret untuk *pan* area grafik.
3.  **Selection (Pilihan):** Klik dan seret di area grafik untuk memilih sekumpulan titik. Titik-titik yang dipilih akan disorot, sementara yang lain mungkin memudar. Ini sangat berguna untuk fokus pada subset data tertentu.
4.  **Warna berdasarkan Kategori:** Warna setiap titik menunjukkan `kategori` produk, yang membantu mengidentifikasi apakah ada pola hubungan yang berbeda antar kategori.

Dari grafik ini, Anda dapat:
*   Melihat apakah ada hubungan linier atau non-linier antara `jumlah` barang yang dibeli dan `total_transaksi`.
*   Mengidentifikasi kelompok-kelompok data berdasarkan `kategori`.
*   Mendeteksi *outlier* atau titik-titik data yang unik yang mungkin menunjukkan transaksi yang tidak biasa.
