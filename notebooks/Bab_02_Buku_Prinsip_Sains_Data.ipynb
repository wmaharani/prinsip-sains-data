## Penggunaan Dasar Pustaka `requests`

Kode ini berfungsi untuk mengambil data dari API publik menggunakan pustaka `requests` di Python. Berikut penjelasannya:

1.  **`import requests`**: Baris ini mengimpor pustaka `requests`, yang digunakan untuk membuat permintaan HTTP (seperti mengambil data dari web).
2.  **`public_api_url = 'https://jsonplaceholder.typicode.com/posts/1'`**: Ini mendefinisikan variabel `public_api_url` yang menyimpan URL dari API publik yang akan diakses. Dalam kasus ini, ini adalah endpoint dari JSONPlaceholder yang menyediakan data *post* sebagai contoh.
3.  **`print(f"Mencoba mengambil data dari: {public_api_url}")`**: Mencetak pesan ke konsol yang menunjukkan URL mana yang sedang dicoba diakses.
4.  **`try...except` blok**: Ini adalah blok penanganan kesalahan. Kode di dalam `try` akan dieksekusi, dan jika ada kesalahan (`requests.exceptions.RequestException`), kode di dalam `except` akan dijalankan.
5.  **`response = requests.get(public_api_url)`**: Ini adalah bagian inti. Ia membuat permintaan HTTP `GET` ke `public_api_url` dan menyimpan responsnya dalam objek `response`.
6.  **`response.raise_for_status()`**: Metode ini memeriksa apakah permintaan berhasil (kode status 200-an). Jika respons menunjukkan kesalahan HTTP (misalnya, 4xx untuk kesalahan klien atau 5xx untuk kesalahan server), maka ia akan memunculkan pengecualian.
7.  **`print(f"Kode Status: {response.status_code}")`**: Mencetak kode status HTTP yang diterima dari server (misalnya, 200 untuk OK).
8.  **`print("JSON Respons:")`** dan **`display(response.json())`**: Jika permintaan berhasil, ini akan mengambil konten respons dalam format JSON (`response.json()`) dan menampilkannya dalam format yang mudah dibaca menggunakan `display()`.
9.  **`except requests.exceptions.RequestException as e: print(f"Terjadi kesalahan: {e}")`**: Jika ada masalah dengan permintaan (misalnya, masalah jaringan, URL salah, atau kesalahan HTTP yang ditangani oleh `raise_for_status`), pesan kesalahan akan dicetak ke konsol.

Secara keseluruhan, kode ini adalah contoh dasar untuk mengambil dan menampilkan data dari API RESTful menggunakan Python.

````python
import requests

# API publik (JSONPlaceholder untuk post)
public_api_url = 'https://jsonplaceholder.typicode.com/posts/1'

print(f"Mencoba mengambil data dari: {public_api_url}")

try:
    response = requests.get(public_api_url)
    response.raise_for_status()  #Pengecualian untuk kesalahan HTTP (4xx atau 5xx)

    print(f"Kode Status: {response.status_code}")
    print("JSON Respons:")
    display(response.json()) # Tampilkan konten JSON dengan baik

except requests.exceptions.RequestException as e:
    print(f"Terjadi kesalahan: {e}")
````

## Memahami dan Mengambil `robots.txt`

Berkas `robots.txt` adalah berkas teks yang diletakkan di direktori akar situs web. Berkas ini digunakan untuk memberikan instruksi kepada *web crawler* (seperti Googlebot) tentang bagian mana dari situs Anda yang boleh atau tidak boleh di-crawl atau diindeks. Ini penting untuk mengelola lalu lintas crawler dan mencegah halaman tertentu muncul di hasil pencarian.

Berikut adalah contoh sederhana dari berkas `robots.txt`:

```
User-agent: *
Disallow: /admin/
Disallow: /private/

User-agent: Googlebot
Allow: /public/
Disallow: /temp/

Sitemap: https://www.contoh-situs.com/sitemap.xml
```

### Penjelasan:

*   **`User-agent: *`**: Aturan ini berlaku untuk semua *web crawler* (robot). Tanda bintang (`*`) adalah *wildcard*.
*   **`Disallow: /admin/`**: Memberi tahu semua robot untuk tidak meng-*crawl* atau mengindeks direktori `/admin/` beserta semua konten di dalamnya.
*   **`Disallow: /private/`**: Sama seperti di atas, melarang akses ke direktori `/private/`.
*   **`User-agent: Googlebot`**: Aturan ini hanya berlaku untuk *crawler* Google (Googlebot).
*   **`Allow: /public/`**: Secara spesifik mengizinkan Googlebot untuk meng-*crawl* direktori `/public/`, bahkan jika ada aturan `Disallow` yang lebih umum yang mungkin mempengaruhinya (meskipun dalam contoh ini tidak ada konflik langsung, `Allow` dapat menimpa `Disallow` yang lebih luas).
*   **`Disallow: /temp/`**: Melarang Googlebot meng-*crawl* direktori `/temp/`.
*   **`Sitemap: https://www.contoh-situs.com/sitemap.xml`**: Baris ini menunjukkan lokasi berkas *sitemap* situs Anda, yang membantu *search engine* menemukan semua halaman penting di situs Anda.

### Mengambil Konten `robots.txt` dengan `requests`

Kode berikut akan menunjukkan cara menggunakan pustaka `requests` untuk mengambil konten berkas `robots.txt` dari suatu situs web. Dalam contoh ini, kita akan mencoba mengambil `robots.txt` dari Google.

````python
import requests

# Tentukan URL dasar situs web
base_url = 'https://www.google.com/'

# Buat URL lengkap untuk berkas robots.txt
robots_txt_url = base_url + 'robots.txt'

print(f"Mencoba mengambil berkas robots.txt dari: {robots_txt_url}")

try:
    # Lakukan permintaan GET ke URL robots.txt
    response = requests.get(robots_txt_url)
    response.raise_for_status()  # Memunculkan pengecualian untuk kode status HTTP yang buruk (4xx atau 5xx)

    print(f"Kode Status: {response.status_code}")
    print("Isi berkas robots.txt:")
    # Cetak konten teks dari respons
    print(response.text)

except requests.exceptions.RequestException as e:
    print(f"Terjadi kesalahan saat mengambil robots.txt: {e}")
    print("Pastikan URL situs web sudah benar dan dapat diakses.")
````

## Contoh Penggunaan Kunci API

Ada dua cara umum untuk mengirimkan kunci API:

1.  **Sebagai Header Otentikasi (`Authorization`):** Ini adalah metode yang paling aman dan direkomendasikan.
2.  **Sebagai Parameter Kueri:** Ini kurang aman karena kunci API mungkin terekam di log server atau histori browser.

````python
import requests

# Ganti dengan kunci API Anda
MY_API_KEY = 'KUNCI_API_ANDA'

# Contoh 1: Mengirim Kunci API sebagai Header Authorization (Paling Direkomendasikan)
# Ini adalah endpoint API yang memerlukan kunci di header
api_url_header = 'https://api.contoh-layanan.com/v1/data'

headers_with_api_key = {
    'Authorization': f'Bearer {MY_API_KEY}', # 'Bearer' adalah skema otentikasi umum
    'Content-Type': 'application/json',
}

print(f"\nMencoba mengambil data dari: {api_url_header} menggunakan header API key")

try:
    response_header = requests.get(api_url_header, headers=headers_with_api_key)
    response_header.raise_for_status() # Tangani kesalahan HTTP

    print(f"Status Code (Header API): {response_header.status_code}")
    # Di sini Anda akan menampilkan data yang sebenarnya jika API ini valid
    # print("Response JSON (Header API):")
    # display(response_header.json())

except requests.exceptions.RequestException as e:
    print(f"Terjadi kesalahan saat menggunakan header API key: {e}")
    print("Pastikan API key dan URL sudah benar dan API tersebut aktif.")
````

````python
import requests

# Ganti dengan kunci API Anda
MY_API_KEY = 'KUNCI_API_ANDA'

# Contoh 2: Mengirim Kunci API sebagai Parameter Kueri
# Ini adalah endpoint API yang memerlukan kunci sebagai parameter URL
api_url_query = 'https://api.contoh-layanan.com/v1/data_publik'

params_with_api_key = {
    'api_key': MY_API_KEY
}

print(f"\nMencoba mengambil data dari: {api_url_query} menggunakan parameter kueri API key")

try:
    response_query = requests.get(api_url_query, params=params_with_api_key)
    response_query.raise_for_status() # Tangani kesalahan HTTP

    print(f"Status Code (Query API): {response_query.status_code}")
    # Di sini Anda akan menampilkan data yang sebenarnya jika API ini valid
    # print("Response JSON (Query API):")
    # display(response_query.json())

except requests.exceptions.RequestException as e:
    print(f"Terjadi kesalahan saat menggunakan parameter kueri API key: {e}")
    print("Pastikan API key dan URL sudah benar dan API tersebut aktif.")
````

Penting: Kedua contoh di atas menggunakan URL dan kunci API hipotetis. Anda perlu mengganti `MY_API_KEY` dengan kunci API Anda yang sebenarnya dan `https://api.contoh-layanan.com/...` dengan titik akhir API nyata yang ingin Anda gunakan.

## Scraping Data dengan `BeautifulSoup`

`BeautifulSoup` adalah pustaka Python untuk mengekstrak data dari berkas HTML dan XML. Ini sangat berguna untuk *web scraping* karena memungkinkan Anda menavigasi, mencari, dan memodifikasi *parse tree*.

### Konsep Utama:

*   **`BeautifulSoup(html_doc, 'html.parser')`**: Menginisialisasi objek `BeautifulSoup`. Anda memberikan dokumen HTML (biasanya dari respons `requests`) dan *parser* yang akan digunakan (misalnya, `'html.parser'` bawaan Python).
*   **Objek `soup`**: Setelah diinisialisasi, objek `soup` mewakili dokumen sebagai struktur pohon, yang dapat Anda navigasikan.
*   **`soup.select('CSS_SELECTOR')`**: Ini adalah metode yang sangat kuat untuk menemukan elemen-elemen HTML menggunakan selektor CSS. Misalnya:
    *   `soup.select('a')`: Mencari semua tag `<a>`.
    *   `soup.select('.kelas-saya')`: Mencari semua elemen dengan kelas `kelas-saya`.
    *   `soup.select('#id-saya')`: Mencari elemen dengan ID `id-saya`.
    *   `soup.select('div.item a')`: Mencari semua tag `<a>` di dalam `<div>` dengan kelas `item`.
*   **`element.get_text(strip=True)`**: Mengambil teks di dalam elemen HTML dan menghapus spasi kosong di awal/akhir.
*   **`element.get('attribute_name', '')`**: Mengambil nilai atribut dari elemen. Misalnya, `tautan.get('href', '')` akan mengambil nilai atribut `href` dari tag `<a>`.

````python
import requests
from bs4 import BeautifulSoup
import time
import pandas as pd

def scrape_buku(url, max_item=10):
    """
    Contoh scraping judul, harga, dan tautan buku dari books.toscrape.com.
    """
    hasil = []
    headers = {
        'User-Agent': 'Mozilla/5.0 (Riset Akademik; contact: nama@email.com)',
    }

    print(f"Mencoba mengambil data dari: {url}")
    try:
        response = requests.get(url, headers=headers, timeout=10)
        response.raise_for_status() # Memunculkan pengecualian untuk kode status HTTP yang buruk (4xx atau 5xx)

        # Parse dokumen HTML
        soup = BeautifulSoup(response.text, 'html.parser')

        # Cari semua 'kartu' buku (sesuaikan selektor CSS dengan situs target)
        # Di books.toscrape.com, setiap buku ada di <article class="product_pod">
        kartu_item = soup.select('article.product_pod')[:max_item]

        for kartu in kartu_item:
            judul_element = kartu.select_one('h3 a')
            harga_element = kartu.select_one('p.price_color')
            tautan_element = kartu.select_one('h3 a')

            if judul_element and harga_element and tautan_element:
                hasil.append({
                    'judul': judul_element.get('title', '').strip(),
                    'harga': harga_element.get_text(strip=True),
                    'tautan': url + tautan_element.get('href', ''), # URL relatif, gabungkan dengan base URL
                })
            time.sleep(0.5)  # Rate limiting: jeda 0.5 detik antar pemrosesan

    except requests.exceptions.RequestException as e:
        print(f"Terjadi kesalahan saat melakukan scraping: {e}")
        return pd.DataFrame() # Kembalikan DataFrame kosong jika terjadi error

    return pd.DataFrame(hasil)

# Penggunaan fungsi
base_url_buku = 'https://books.toscrape.com/'
df_buku = scrape_buku(base_url_buku, max_item=5) # Ambil 5 buku pertama

if not df_buku.empty:
    print("\nHasil scraping:")
    display(df_buku)
else:
    print("Tidak ada data yang berhasil discrape.")
````

Berikut adalah contoh bagaimana kita dapat membuat dan membaca berkas JSON dengan struktur bersarang.

````python
import pandas as pd
import json

# Membuat data dummy dengan struktur bersarang untuk JSON
nested_json_data = [
    {
        "order_id": "ORD001",
        "customer": {
            "id": "CUST001",
            "name": "Alice",
            "email": "alice@example.com"
        },
        "items": [
            {"item_id": "ITEM001", "product_name": "Laptop", "quantity": 1, "price": 1200},
            {"item_id": "ITEM002", "product_name": "Mouse", "quantity": 2, "price": 25}
        ],
        "total_amount": 1250
    },
    {
        "order_id": "ORD002",
        "customer": {
            "id": "CUST002",
            "name": "Bob",
            "email": "bob@example.com"
        },
        "items": [
            {"item_id": "ITEM003", "product_name": "Keyboard", "quantity": 1, "price": 75}
        ],
        "total_amount": 75
    }
]

# Menyimpan data dummy bersarang ke file JSON
nested_json_file_path = "sample_data/contoh_dummy_nested.json"
with open(nested_json_file_path, 'w') as f:
    json.dump(nested_json_data, f, indent=4)

print(f"Berkas JSON dummy bersarang '{nested_json_file_path}' berhasil dibuat.")
````

## Mengambil Data dari BPS API (Badan Pusat Statistik)

Kode ini menunjukkan cara mengambil data *time series* dari API Badan Pusat Statistik (BPS) menggunakan `requests` dan mengolahnya dengan `pandas`. Ini adalah contoh penggunaan API otentikasi yang lebih kompleks, di mana Anda memerlukan kunci API.

### Penjelasan Kode:

1.  **Impor Pustaka**: Mengimpor `requests` untuk permintaan HTTP, `pandas` untuk manipulasi data, `os` untuk berinteraksi dengan sistem operasi, dan `dotenv` untuk memuat variabel lingkungan.

2.  **Memuat Kunci API (`.env` file)**:
    *   `load_dotenv()`: Fungsi ini mencari berkas `.env` di direktori proyek Anda dan memuat variabel-variabel yang didefinisikan di dalamnya ke dalam variabel lingkungan sistem.
    *   `API_KEY = os.getenv('BPS_API_KEY')`: Mengambil nilai kunci API dari variabel lingkungan bernama `BPS_API_KEY`. Ini adalah praktik terbaik untuk menyimpan kunci API di luar kode Anda (misalnya, di berkas `.env`) untuk alasan keamanan dan agar tidak langsung terlihat di kode yang Anda bagikan.

3.  **Fungsi `ambil_data_bps`**:
    *   **Tujuan**: Fungsi ini dirancang untuk mengambil data BPS berdasarkan ID variabel (`var_id`), rentang tahun, dan kunci API.
    *   **`base_url` dan `endpoint`**: Mendefinisikan URL dasar API BPS dan membangun `endpoint` permintaan dengan parameter seperti `lang` (bahasa), `domain` (kode domain statistik), `var` (variabel yang diminta), `tahun` (tahun awal), dan `key` (kunci API Anda).
    *   **Permintaan HTTP**: `requests.get(endpoint, timeout=15)` membuat permintaan GET ke API BPS dengan batas waktu 15 detik. Ini penting untuk mencegah permintaan menggantung tanpa batas jika server tidak merespons.
    *   **Penanganan Status Respons**: `response.raise_for_status()` akan memunculkan `HTTPError` jika permintaan gagal (misalnya, status 4xx atau 5xx).
    *   **Ekstraksi Data**: Jika respons berhasil (`status == 'OK'`), kode mengambil `datacontent` dari JSON respons. `datacontent` adalah objek yang berisi data, di mana kuncinya adalah kode dan nilainya adalah data yang sebenarnya.
    *   **Pembentukan DataFrame**: `pd.DataFrame(records.items(), columns=['kode', 'nilai'])` mengubah data yang diekstrak menjadi DataFrame `pandas` dengan kolom 'kode' dan 'nilai'.
    *   **Konversi Tipe Data**: `df['nilai'] = pd.to_numeric(df['nilai'], errors='coerce')` mengonversi kolom 'nilai' menjadi numerik. `errors='coerce'` akan mengubah nilai yang tidak bisa dikonversi menjadi `NaN` (Not a Number), mencegah kesalahan.
    *   **Penanganan Kesalahan API**: Jika `status` dari respons JSON bukan 'OK', fungsi akan memunculkan `ValueError` dengan pesan kesalahan dari API.

4.  **Contoh Penggunaan (dikomentari)**:
    *   `df_inflasi = ambil_data_bps(var_id=1734)`: Ini adalah contoh bagaimana Anda memanggil fungsi untuk mengambil data inflasi (dengan asumsi `var_id=1734` adalah ID variabel untuk inflasi di BPS).
    *   `print(df_inflasi.head())`: Menampilkan beberapa baris pertama data yang diambil.
    *   `df_inflasi.to_csv('data_inflasi_bps.csv', index=False)`: Menyimpan data yang diambil ke berkas CSV lokal. `index=False` mencegah indeks DataFrame ditulis sebagai kolom di CSV.

**Penting**: Sebelum menggunakan kode ini, pastikan Anda telah membuat berkas `.env` di folder yang sama dengan notebook Anda, dan di dalamnya terdapat baris `BPS_API_KEY=KunciAPIAndaYangSebenarnya` (ganti `KunciAPIAndaYangSebenarnya` dengan kunci API BPS yang valid). Anda dapat memperoleh kunci API BPS dari situs web resmi mereka.

````python
import requests
import pandas as pd
import os
from dotenv import load_dotenv

# Muat API Key dari file .env
load_dotenv()
API_KEY = os.getenv('BPS_API_KEY')

def ambil_data_bps(var_id, tahun_mulai=2020, tahun_akhir=2024):
    "Mengambil data time series dari API BPS."
    base_url = 'https://webapi.bps.go.id/v1/api'
    endpoint = f'{base_url}/list/model/data/lang/ind/domain/0000'
    endpoint += f'/var/{var_id}/tahun/{tahun_mulai}/key/{API_KEY}'

    response = requests.get(endpoint, timeout=15)
    response.raise_for_status()
    data_json = response.json()

    # Ekstrak data dari struktur JSON
    if data_json.get('status') == 'OK':
        records = data_json.get('datacontent', {})
        df = pd.DataFrame(records.items(), columns=['kode', 'nilai'])
        df['nilai'] = pd.to_numeric(df['nilai'], errors='coerce')
        return df
    else:
        raise ValueError(f'API Error: {data_json.get("message")}')

# Contoh penggunaan
# df_inflasi = ambil_data_bps(var_id=1734)
# print(df_inflasi.head())
# df_inflasi.to_csv('data_inflasi_bps.csv', index=False)
````
