Setelah melihat ringkasan statistik deskriptif umum, kita akan menghitung Ukuran Tendensi Sentral secara spesifik untuk setiap kolom numerik yang telah kita definisikan sebelumnya (`kolom_numerik`). Ukuran tendensi sentral meliputi:

*   **Mean (Rata-rata)**: Jumlah semua nilai dibagi dengan jumlah observasi.
*   **Median**: Nilai tengah dalam serangkaian observasi yang diurutkan.
*   **Modus**: Nilai yang paling sering muncul dalam serangkaian observasi.

Perhitungan ini akan membantu kita memahami di mana sebagian besar data terkonsentrasi dan apakah ada perbedaan signifikan antara mean dan median (yang bisa mengindikasikan adanya pencilan atau distribusi yang miring).

```python
print("Ukuran Tendensi Sentral untuk Kolom Numerik:")
for col in kolom_numerik:
    mean_val = df[col].mean()
    median_val = df[col].median()
    mode_val = df[col].mode().tolist()

    print(f"\nKolom: {col}")
    print(f"  Mean: {mean_val:.2f}")
    print(f"  Median: {median_val:.2f}")
    print(f"  Mode: {mode_val}")
```

Hasil di atas menunjukkan ukuran dispersi untuk kolom-kolom numerik yang kita pilih (`total_rooms`, `population`, `median_income`, `median_house_value`). Ukuran dispersi ini membantu kita memahami seberapa bervariasi atau tersebarnya data dalam setiap kolom:

*   **Varians**: Mengukur seberapa jauh setiap angka dalam kumpulan data dari rata-rata. Nilai varians yang tinggi menunjukkan data sangat tersebar.
*   **Standar Deviasi**: Akar kuadrat dari varians. Ini adalah ukuran sebaran data yang paling umum digunakan dan lebih mudah diinterpretasikan karena memiliki satuan yang sama dengan data aslinya. Semakin besar standar deviasi, semakin lebar sebaran data.
*   **IQR (Interquartile Range)**: Merupakan rentang antara kuartil ketiga (75%) dan kuartil pertama (25%). IQR tidak terpengaruh oleh pencilan (outlier) ekstrem dan menunjukkan sebaran 50% data di tengah.
*   **MAD (Mean Absolute Deviation)**: Rata-rata dari selisih absolut antara setiap titik data dan rata-rata. MAD juga kurang sensitif terhadap pencilan dibandingkan standar deviasi.
*   **Rentang (Range)**: Perbedaan antara nilai maksimum dan minimum dalam kumpulan data. Ini adalah ukuran dispersi yang paling sederhana tetapi sangat sensitif terhadap pencilan.
*   **Koefisien Variasi**: Mengukur sebaran relatif data terhadap rata-ratanya. Ini berguna untuk membandingkan tingkat variasi antara kumpulan data yang memiliki rata-rata yang sangat berbeda. Dinyatakan dalam persentase.

Dari hasil ini, kita bisa melihat bahwa `total_rooms` dan `population` memiliki varians dan standar deviasi yang sangat besar, menunjukkan variasi yang tinggi dalam jumlah ruangan dan populasi antar lokasi. `median_income` memiliki varians dan standar deviasi yang relatif kecil, menunjukkan data pendapatan median cenderung lebih terkonsentrasi. `median_house_value` juga memiliki variasi yang signifikan, namun koefisien variasinya menunjukkan sebaran relatif terhadap rata-rata yang perlu dipertimbangkan.

```python
print("Ukuran Dispersi untuk Kolom Numerik:")
for col in kolom_numerik:
    varians = df[col].var()
    std_deviasi = df[col].std()
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    iqr = Q3 - Q1
    mad = (df[col] - df[col].mean()).abs().mean()
    rentang = df[col].max() - df[col].min()
    koefisien_variasi = (std_deviasi / df[col].mean()) * 100 if df[col].mean() != 0 else 0

    print(f"\nKolom: {col}")
    print(f"  Varians: {varians:.2f}")
    print(f"  Standar Deviasi: {std_deviasi:.2f}")
    print(f"  IQR (Interquartile Range): {iqr:.2f}")
    print(f"  MAD (Mean Absolute Deviation): {mad:.2f}")
    print(f"  Rentang (Range): {rentang:.2f}")
    print(f"  Koefisien Variasi: {koefisien_variasi:.2f}%")
```

Hasil di atas memberikan gambaran tentang bentuk distribusi data untuk setiap kolom numerik:

*   **Skewness (Kemiringan)**: Mengukur asimetri distribusi data. Jika:
    *   **Skewness ≈ 0**: Distribusi relatif simetris.
    *   **Skewness > 0 (Positif)**: Distribusi miring ke kanan (tail lebih panjang di sisi kanan), menunjukkan adanya nilai-nilai tinggi yang ekstrem (pencilan kanan).
    *   **Skewness < 0 (Negatif)**: Distribusi miring ke kiri (tail lebih panjang di sisi kiri), menunjukkan adanya nilai-nilai rendah yang ekstrem (pencilan kiri).

*   **Kurtosis (Keruncingan)**: Mengukur 'keruncingan' atau 'ketinggian' puncak distribusi dan 'ketebalan' ekornya dibandingkan dengan distribusi normal (yang memiliki kurtosis sekitar 3, atau 0 dalam definisi 'excess kurtosis' yang digunakan Pandas).
    *   **Kurtosis ≈ 0 (Mesokurtic)**: Bentuk distribusi mirip dengan distribusi normal.
    *   **Kurtosis > 0 (Leptokurtic)**: Distribusi lebih 'runcing' dengan ekor yang lebih 'berat' (lebih banyak pencilan ekstrem) dibandingkan distribusi normal.
    *   **Kurtosis < 0 (Platykurtic)**: Distribusi lebih 'datar' dengan ekor yang lebih 'ringan' (lebih sedikit pencilan ekstrem) dibandingkan distribusi normal.

Dari hasil:
*   `total_rooms`, `population`, dan `median_income` memiliki **skewness positif yang tinggi**, menunjukkan bahwa data untuk kolom-kolom ini sangat miring ke kanan. Ini berarti ada banyak properti dengan jumlah kamar, populasi, dan pendapatan median yang relatif rendah, namun ada juga sejumlah kecil properti dengan nilai yang sangat tinggi.
*   `median_house_value` juga memiliki **skewness positif** (0.97), namun tidak setinggi kolom lainnya, menunjukkan kemiringan ke kanan yang lebih moderat.
*   Semua kolom menunjukkan **kurtosis positif yang tinggi** (terutama `total_rooms` dan `population`), yang mengindikasikan bahwa distribusi ini memiliki puncak yang lebih runcing dan ekor yang lebih 'berat' daripada distribusi normal. Ini berarti ada lebih banyak pencilan ekstrem di kedua sisi distribusi dibandingkan dengan distribusi normal, atau data sangat terkonsentrasi di sekitar rata-rata dengan beberapa nilai ekstrem yang menyebar jauh.

```python
print("Ukuran Bentuk Distribusi (Skewness dan Kurtosis) untuk Kolom Numerik:")
for col in kolom_numerik:
    skewness = df[col].skew()
    kurtosis = df[col].kurt()

    print(f"\nKolom: {col}")
    print(f"  Skewness: {skewness:.2f}")
    print(f"  Kurtosis: {kurtosis:.2f}")
```

Visualisasi histogram di atas menunjukkan distribusi setiap kolom numerik (`total_rooms`, `population`, `median_income`, dan `median_house_value`) menggunakan berbagai strategi pemilihan lebar bin.

Memilih lebar bin yang tepat sangat penting dalam histogram karena dapat sangat memengaruhi interpretasi bentuk distribusi data:

*   **Bin Sempit (misal: 100 bins)**:
    *   Menghasilkan tampilan yang sangat detail dari data, memperlihatkan setiap 'gigi' atau fluktuasi kecil. Ini bisa sangat berguna untuk mengidentifikasi mode lokal atau celah dalam data yang mungkin tidak terlihat dengan bin yang lebih lebar.
    *   Namun, bin yang terlalu sempit juga bisa membuat histogram terlihat 'berisik' dan sulit untuk melihat pola umum distribusi, serta mungkin sensitif terhadap pencilan (outlier).

*   **Bin Lebar (misal: 10 bins)**:
    *   Menghasilkan tampilan yang lebih halus dan umum dari distribusi. Ini baik untuk melihat tren dasar atau bentuk keseluruhan distribusi (misalnya, apakah miring ke kanan atau kiri).
    *   Namun, bin yang terlalu lebar dapat menyembunyikan detail penting, seperti keberadaan beberapa mode (multimodality) atau celah kecil dalam data.

*   **Bin Otomatis (`'auto'`)**: Algoritma yang digunakan oleh `seaborn`/`matplotlib` untuk secara heuristik memilih jumlah bin yang 'terbaik' berdasarkan karakteristik data. Tujuannya adalah menyeimbangkan antara terlalu banyak detail (bin sempit) dan terlalu sedikit detail (bin lebar).

*   **Algoritma Bin Khusus (Sturges, Scott, Freedman-Diaconis)**:
    *   **Sturges**: Sebuah aturan yang cukup tua dan sering digunakan. Cenderung menghasilkan jumlah bin yang relatif kecil, cocok untuk distribusi data yang hampir normal.
    *   **Scott**: Berdasarkan estimasi standar deviasi. Cenderung menghasilkan bin yang lebih lebar dan dapat bekerja dengan baik untuk distribusi yang mendekati normal.
    *   **Freedman-Diaconis (FD)**: Metode yang lebih robust terhadap pencilan (outlier) karena menggunakan rentang interkuartil (IQR) untuk menentukan lebar bin. Ini sering kali menghasilkan bin yang lebih sempit daripada Sturges atau Scott dan lebih baik dalam mengungkapkan detail distribusi ketika ada banyak pencilan.

**Observasi dari Histogram:**

Dari visualisasi yang ditampilkan, kita dapat melihat bagaimana perbedaan lebar bin memengaruhi representasi distribusi:

*   Untuk kolom seperti `total_rooms` dan `population` (yang memiliki skewness dan kurtosis tinggi), **bin sempit** dan algoritma seperti **Freedman-Diaconis** sering kali menunjukkan distribusi yang sangat terkonsentrasi pada nilai-nilai rendah dengan 'ekor' panjang ke kanan, yang mengkonfirmasi adanya pencilan (outlier) besar. Bin lebar cenderung menyembunyikan detail ini dan membuat distribusi tampak lebih 'normal' atau sederhana.
*   Pemilihan bin 'otomatis' sering kali memberikan kompromi yang baik, namun tergantung pada karakteristik data, algoritma khusus (seperti FD untuk data dengan pencilan) bisa lebih informatif.

Secara keseluruhan, dengan membandingkan histogram dengan bin yang berbeda, kita mendapatkan pemahaman yang lebih kaya tentang struktur dan karakteristik setiap variabel numerik kontinu dalam dataset.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Memastikan kolom_numerik sudah terdefinisi dari konteks sebelumnya
if 'kolom_numerik' not in locals() or not kolom_numerik:
    kolom_numerik = df.select_dtypes(include=np.number).columns.tolist()
    # Hapus kolom yang bukan continuous jika ada
    if 'housing_median_age' in kolom_numerik: # Contoh jika ingin mengecualikan housing_median_age
        kolom_numerik.remove('housing_median_age')
    print(f"Kolom numerik yang akan divisualisasikan: {kolom_numerik}")

print("Visualisasi Histogram untuk Variabel Numerik Kontinu dengan Berbagai Lebar Bin:\n")

for col in kolom_numerik:
    print(f"Visualisasi untuk Kolom: {col}")
    fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(15, 15))
    fig.suptitle(f'Histogram untuk {col} dengan Berbagai Pilihan Bins', fontsize=16)

    # Bin Sempit (misal: 100 bins)
    sns.histplot(df[col], bins=100, kde=True, ax=axes[0, 0])
    axes[0, 0].set_title(f'Bin Sempit (100 bins)')
    axes[0, 0].set_xlabel(col)
    axes[0, 0].set_ylabel('Frekuensi')

    # Bin Lebar (misal: 10 bins)
    sns.histplot(df[col], bins=10, kde=True, ax=axes[0, 1])
    axes[0, 1].set_title(f'Bin Lebar (10 bins)')
    axes[0, 1].set_xlabel(col)
    axes[0, 1].set_ylabel('Frekuensi')

    # Auto Bins (default algorithm by seaborn/matplotlib)
    sns.histplot(df[col], bins='auto', kde=True, ax=axes[1, 0])
    axes[1, 0].set_title(f'Bin Otomatis (bins=\'auto\')')
    axes[1, 0].set_xlabel(col)
    axes[1, 0].set_ylabel('Frekuensi')

    # Sturges Bins
    sns.histplot(df[col], bins='sturges', kde=True, ax=axes[1, 1])
    axes[1, 1].set_title(f'Bin Sturges (bins=\'sturges\')')
    axes[1, 1].set_xlabel(col)
    axes[1, 1].set_ylabel('Frekuensi')

    # Scott Bins
    sns.histplot(df[col], bins='scott', kde=True, ax=axes[2, 0])
    axes[2, 0].set_title(f'Bin Scott (bins=\'scott\')')
    axes[2, 0].set_xlabel(col)
    axes[2, 0].set_ylabel('Frekuensi')

    # Freedman-Diaconis Bins
    sns.histplot(df[col], bins='fd', kde=True, ax=axes[2, 1])
    axes[2, 1].set_title(f'Bin Freedman-Diaconis (bins=\'fd\')')
    axes[2, 1].set_xlabel(col)
    axes[2, 1].set_ylabel('Frekuensi')

    plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout to prevent title overlap
    plt.show()
    print("\n" + "="*70 + "\n") # Separator for better readability
```

Visualisasi Histogram memberikan gambaran yang baik tentang frekuensi data dalam setiap bin. Namun, untuk mendapatkan representasi yang lebih halus dari distribusi probabilitas data kontinu, kita dapat menggunakan **Kernel Density Estimation (KDE) Plot**.

### Kernel Density Estimation (KDE) Plot

*   **Apa itu KDE?**
    KDE adalah metode non-parametrik untuk mengestimasi fungsi kepadatan probabilitas (PDF) dari variabel acak kontinu. Ini pada dasarnya "menghaluskan" histogram dengan menempatkan fungsi "kernel" (biasanya Gaussian) di setiap titik data dan menjumlahkan fungsi-fungsi ini untuk membuat estimasi kurva densitas.

*   **Bagaimana cara kerjanya?**
    Bayangkan setiap titik data memiliki sebuah "gundukan" kecil (kernel) di sekitarnya. KDE akan menjumlahkan semua gundukan ini untuk membentuk sebuah kurva yang mulus. Area di bawah kurva KDE selalu sama dengan 1, sama seperti fungsi kepadatan probabilitas. Ini berarti tinggi kurva pada titik tertentu menunjukkan "densitas" atau kemungkinan relatif data berada di sekitar titik tersebut.

*   **Kapan digunakan?**
    KDE sangat berguna untuk:
    *   **Mengidentifikasi bentuk distribusi**: Lebih jelas menunjukkan kemiringan (skewness), puncak (mode), dan ekor distribusi dibandingkan histogram.
    *   **Membandingkan distribusi**: Mudah untuk memvisualisasikan perbedaan antara distribusi dua atau lebih kelompok data.
    *   **Menghindari masalah binning**: Tidak seperti histogram, KDE tidak bergantung pada pemilihan lebar atau lokasi bin, sehingga memberikan representasi yang lebih objektif.

*   **Bandwidth**: Ini adalah parameter kunci dalam KDE, mirip dengan lebar bin pada histogram. Bandwidth yang kecil akan menghasilkan kurva yang "bergerigi" dan mungkin overfit data, sementara bandwidth yang besar akan menghasilkan kurva yang terlalu halus dan mungkin menyembunyikan detail penting. Seaborn dan Matplotlib biasanya memilih bandwidth secara otomatis berdasarkan algoritma yang baik.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Memastikan kolom_numerik sudah terdefinisi dari konteks sebelumnya
if 'kolom_numerik' not in locals() or not kolom_numerik:
    kolom_numerik = df.select_dtypes(include=np.number).columns.tolist()
    # Hapus kolom yang bukan continuous jika ada
    if 'housing_median_age' in kolom_numerik:
        kolom_numerik.remove('housing_median_age')
    print(f"Kolom numerik yang akan divisualisasikan: {kolom_numerik}")

print("Visualisasi Kernel Density Estimation (KDE) untuk Variabel Numerik Kontinu:\n")

plt.figure(figsize=(15, len(kolom_numerik) * 5))
for i, col in enumerate(kolom_numerik):
    plt.subplot(len(kolom_numerik), 1, i + 1)
    sns.kdeplot(df[col], fill=True, color='skyblue')
    plt.title(f'KDE Plot untuk {col}', fontsize=14)
    plt.xlabel(col)
    plt.ylabel('Densitas')
    plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

Setelah melihat distribusi data melalui histogram dan KDE plot, kita akan menggunakan **Boxplot** untuk mendapatkan ringkasan visual yang cepat tentang pusat, sebaran, dan keberadaan *outlier* (pencilan) dalam data numerik kontinu.

### Boxplot

Boxplot adalah metode standar untuk menampilkan distribusi data berdasarkan ringkasan lima angka (five-number summary):

1.  **Nilai Minimum**: Titik terendah data yang bukan *outlier*.
2.  **Kuartil Pertama (Q1)**: 25% data berada di bawah nilai ini. Ini adalah batas bawah kotak.
3.  **Median (Q2)**: Nilai tengah data (50% data di bawahnya). Ini adalah garis di dalam kotak.
4.  **Kuartil Ketiga (Q3)**: 75% data berada di bawah nilai ini. Ini adalah batas atas kotak.
5.  **Nilai Maksimum**: Titik tertinggi data yang bukan *outlier*.

Selain itu, boxplot juga secara eksplisit menunjukkan:

*   **Kotak (Box)**: Merentang dari Q1 hingga Q3, mewakili 50% data tengah atau *Interquartile Range* (IQR = Q3 - Q1).
*   **Whiskers**: Memanjang dari tepi kotak hingga nilai terkecil/terbesar dalam rentang 1.5 kali IQR dari tepi kotak. Nilai-nilai di luar rentang ini dianggap sebagai *outlier*.
*   **Titik (*Outliers*)**: Titik-titik individual di luar *whiskers* menunjukkan *outlier* potensial yang mungkin memerlukan penyelidikan lebih lanjut.

**Mengapa Boxplot Penting?**

*   **Ringkasan**: Memberikan gambaran singkat tentang lokasi, sebaran, dan kemiringan data.
*   **Deteksi Outlier**: Sangat efektif dalam mengidentifikasi nilai-nilai ekstrem atau *outlier*.
*   **Perbandingan Distribusi**: Memudahkan perbandingan distribusi antar kelompok atau variabel.

Dengan boxplot, kita dapat dengan mudah melihat di mana sebagian besar data terkonsentrasi, seberapa tersebar datanya, dan apakah ada nilai-nilai yang jauh menyimpang dari pola umum data.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Memastikan kolom_numerik sudah terdefinisi dari konteks sebelumnya
# Jika belum, akan diinisialisasi ulang
if 'kolom_numerik' not in locals() or not kolom_numerik:
    kolom_numerik = df.select_dtypes(include=np.number).columns.tolist()
    if 'housing_median_age' in kolom_numerik:
        kolom_numerik.remove('housing_median_age')
    print(f"Kolom numerik yang akan divisualisasikan: {kolom_numerik}")

print("Visualisasi Boxplot untuk Variabel Numerik Kontinu:\n")

plt.figure(figsize=(15, len(kolom_numerik) * 4))
for i, col in enumerate(kolom_numerik):
    plt.subplot(len(kolom_numerik), 1, i + 1)
    sns.boxplot(x=df[col], color='lightgreen')
    plt.title(f'Boxplot untuk {col}', fontsize=14)
    plt.xlabel(col)
    plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

### Violin Plot

Violin plot adalah cara visualisasi data yang kaya informasi yang menunjukkan baik kepadatan probabilitas data (mirip KDE) maupun ringkasan statistik (mirip boxplot). Ini sangat berguna untuk memahami bentuk distribusi data secara detail, termasuk kepadatan data di berbagai rentang nilai.

*   **Bentuk 'Biola' (Kernel Density Estimation)**:
    *   Bagian terluar dari violin plot membentuk bentuk 'biola' yang mencerminkan fungsi kepadatan probabilitas (KDE) dari data. Lebar 'biola' pada titik tertentu menunjukkan densitas atau frekuensi data di sekitar nilai tersebut. Bagian yang lebih lebar berarti lebih banyak data terkonsentrasi di sana, sementara bagian yang lebih sempit menunjukkan lebih sedikit data.
    *   Ini membantu kita melihat puncak (mode) distribusi, kemiringan (skewness), dan sebaran data secara keseluruhan dengan cara yang intuitif.

*   **Elemen Boxplot di Dalam**:
    *   Secara default, atau dengan menggunakan parameter `inner='quartile'` seperti yang akan kita gunakan, violin plot juga dapat menampilkan elemen-elemen dari boxplot di dalamnya.
    *   Biasanya, garis tebal di tengah mewakili median, dan dua garis tipis di kedua sisi median mewakili kuartil pertama (Q1) dan kuartil ketiga (Q3).
    *   Ini memberikan informasi ringkasan statistik (median, IQR) di atas representasi densitas.

**Keunggulan Violin Plot:**

*   **Informasi Lebih Lengkap**: Violin plot memberikan informasi yang lebih kaya tentang distribusi dibandingkan boxplot, karena tidak hanya menampilkan ringkasan statistik tetapi juga kepadatan data.
*   **Identifikasi Bentuk Distribusi**: Sangat baik untuk melihat bentuk distribusi data, termasuk apakah distribusi itu multimodal (memiliki beberapa puncak).
*   **Perbandingan yang Efektif**: Sangat efektif untuk membandingkan distribusi antara kelompok yang berbeda karena bentuk 'biola' secara visual dapat mengungkapkan perbedaan dalam kepadatan dan sebaran data.

Dengan violin plot, kita dapat dengan jelas melihat di mana sebagian besar data terkonsentrasi, seberapa tersebar datanya, dan apakah ada pola khusus dalam distribusinya, seperti multiple modes atau asimetri yang kuat.

### Eksplorasi Bivariat: Numerik vs Numerik dengan Scatter Plot

Untuk memahami hubungan antara dua variabel numerik kontinu, **Scatter Plot (Plot Sebar)** adalah alat visualisasi yang paling fundamental dan efektif.

*   **Apa itu Scatter Plot?**
    *Scatter plot* adalah grafik yang menampilkan titik-titik data pada sistem koordinat Cartesian, di mana setiap titik merepresentasikan pasangan nilai dari dua variabel yang berbeda.
    *   Satu variabel diplot pada sumbu horizontal (x-axis).
    *   Variabel lainnya diplot pada sumbu vertikal (y-axis).

*   **Tujuan dan Informasi yang Diberikan:**
    1.  **Arah Hubungan (Korelasi)**:
        *   **Positif**: Jika titik-titik cenderung naik dari kiri bawah ke kanan atas, menunjukkan bahwa ketika nilai satu variabel meningkat, nilai variabel lain juga cenderung meningkat.
        *   **Negatif**: Jika titik-titik cenderung turun dari kiri atas ke kanan bawah, menunjukkan bahwa ketika nilai satu variabel meningkat, nilai variabel lain cenderung menurun.
        *   **Tidak Ada Hubungan**: Jika titik-titik tersebar secara acak tanpa pola yang jelas.
    2.  **Kekuatan Hubungan**: Seberapa dekat titik-titik data mengikuti pola tertentu. Titik-titik yang sangat dekat dengan garis lurus menunjukkan hubungan yang kuat.
    3.  **Bentuk Hubungan**: Apakah hubungan itu linier (membentuk garis lurus) atau non-linier (membentuk kurva).
    4.  **Keberadaan Outlier**: Titik-titik data yang jauh dari pola utama dapat mengindikasikan *outlier* atau pencilan.
    5.  **Kepadatan Data**: Area di mana titik-titik data lebih padat menunjukkan konsentrasi observasi yang lebih tinggi untuk kombinasi nilai kedua variabel tersebut.

*   **Contoh Penerapan:**
    Dalam dataset perumahan, hubungan antara `median_income` (pendapatan median) dan `median_house_value` (nilai median rumah) adalah salah satu yang paling sering dianalisis. Kita mengharapkan adanya korelasi positif: semakin tinggi pendapatan di suatu area, semakin tinggi pula nilai properti di sana.

    *Scatter plot* ini akan memungkinkan kita untuk secara visual mengkonfirmasi hipotesis ini dan juga mengidentifikasi nuansa dalam hubungan tersebut, seperti apakah ada batas atas (capping) pada nilai rumah atau kelompok-kelompok yang berbeda dalam data.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Memilih dua kolom numerik untuk scatter plot
x_var = 'median_income'
y_var = 'median_house_value'

plt.figure(figsize=(10, 6))
sns.scatterplot(x=df[x_var], y=df[y_var], alpha=0.6, s=20, color='darkblue')
plt.title(f'Scatter Plot: {y_var} vs {x_var}', fontsize=16)
plt.xlabel(x_var, fontsize=12)
plt.ylabel(y_var, fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Eksplorasi Bivariat: Kategorikal vs Numerik dengan Swarm Plot

Ketika kita memiliki satu variabel kategorikal (atau pseudo-kategorikal seperti `housing_median_age`) dan satu variabel numerik (seperti `median_house_value`), kita ingin melihat bagaimana nilai-nilai numerik tersebut terdistribusi di dalam setiap kelompok kategori.

*   **Mengapa Swarm Plot?**
    *Swarm plot* adalah pilihan yang baik karena ia menampilkan semua titik data individual, menghindari *overplotting* yang sering terjadi pada *strip plot* dengan 'menyebarkan' titik-titik data agar tidak saling tumpang tindih. Ini memungkinkan kita untuk melihat kepadatan data di setiap kategori serta sebaran keseluruhan.

*   **Tujuan dan Informasi yang Diberikan:**
    1.  **Distribusi per Kategori**: Kita dapat melihat secara langsung rentang nilai, konsentrasi data, dan *outlier* untuk variabel numerik di setiap kategori. Misalnya, bagaimana rentang harga rumah berubah untuk setiap usia median rumah.
    2.  **Perbandingan Distribusi**: Memudahkan perbandingan pola distribusi (median, kepadatan, sebaran) dari variabel numerik di antara kategori-kategori yang berbeda.
    3.  **Identifikasi Tren**: Meskipun `housing_median_age` adalah pseudo-kategorikal, kita masih bisa mencari tren: apakah ada korelasi (misalnya, harga rumah cenderung lebih tinggi pada usia rumah tertentu)?

*   **Interpretasi untuk `housing_median_age` dan `median_house_value`:**
    Kita akan menggunakan `housing_median_age` sebagai variabel kategorikal (pada sumbu x) dan `median_house_value` sebagai variabel numerik (pada sumbu y). Plot ini akan menunjukkan:
    *   Bagaimana nilai median rumah (`median_house_value`) tersebar untuk setiap usia median rumah (`housing_median_age`).
    *   Apakah ada kelompok usia rumah tertentu yang cenderung memiliki nilai rumah lebih tinggi atau lebih rendah.
    *   Kepadatan data, yaitu di mana sebagian besar observasi jatuh dalam setiap kategori usia.

### Peringatan Saat Menggunakan Swarm Plot untuk Data Padat:

Sangat penting untuk dicatat bahwa *swarm plot* paling efektif untuk dataset dengan jumlah observasi yang moderat. Untuk dataset yang sangat besar atau kategori dengan banyak titik data, *swarm plot* dapat menghasilkan peringatan seperti:

*   **`UserWarning: X% of the points cannot be placed; you may want to decrease the size of the markers or use stripplot.`**
    Peringatan ini berarti *swarmplot* tidak dapat menempatkan semua titik data individual tanpa tumpang tindih karena kepadatan data yang tinggi. Ini bisa berarti bahwa plot yang dihasilkan tidak sepenuhnya merepresentasikan semua data, dan beberapa informasi mungkin hilang.

*   **`FutureWarning: Passing palette without assigning hue is deprecated...`**
    Ini adalah peringatan umum dari Seaborn yang menunjukkan perubahan perilaku parameter `palette` di masa depan jika `hue` tidak ditetapkan. Untuk tujuan visualisasi dasar, ini tidak memengaruhi hasil plot secara fungsional.

Jika peringatan penempatan titik terlalu sering muncul atau persentase titik yang tidak ditempatkan tinggi, visualisasi alternatif seperti *boxplot* atau *violinplot* mungkin lebih cocok karena mereka merangkum distribusi daripada mencoba menampilkan setiap titik individual.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Memilih variabel pseudo-kategorikal dan numerik
x_cat = 'housing_median_age'
y_num = 'median_house_value'

plt.figure(figsize=(15, 8))
sns.swarmplot(x=df[x_cat], y=df[y_num], palette='viridis', size=3)
plt.title(f'Swarm Plot: {y_num} vs {x_cat}', fontsize=16)
plt.xlabel(f'{x_cat} (Tahun)', fontsize=12)
plt.ylabel(y_num, fontsize=12)
plt.xticks(rotation=45, ha='right') # Rotasi label x agar mudah dibaca
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Analisis Korelasi Pearson: Pemahaman dan Interpretasi

**Korelasi Pearson (Pearson's r)** adalah salah satu ukuran statistik yang paling umum digunakan untuk mengukur kekuatan dan arah hubungan linier antara dua variabel numerik kontinu.

**Rentang Nilai Koefisien Korelasi Pearson:**
Koefisien korelasi Pearson (r) selalu memiliki nilai antara -1 dan +1.

*   **r = +1**: Menunjukkan hubungan linier positif sempurna. Artinya, ketika nilai satu variabel meningkat, nilai variabel lain juga meningkat secara proporsional dan konsisten.
*   **r = -1**: Menunjukkan hubungan linier negatif sempurna. Artinya, ketika nilai satu variabel meningkat, nilai variabel lain akan menurun secara proporsional dan konsisten.
*   **r = 0**: Menunjukkan tidak ada hubungan linier antara kedua variabel. Penting untuk diingat bahwa korelasi nol tidak berarti tidak ada hubungan sama sekali, hanya tidak ada hubungan *linier*.

**Interpretasi Kekuatan Korelasi (Panduan Umum, dapat bervariasi tergantung bidang ilmu):**

*   **0.0 hingga ±0.19**: Korelasi sangat lemah (dapat diabaikan).
*   **±0.20 hingga ±0.39**: Korelasi lemah.
*   **±0.40 hingga ±0.59**: Korelasi moderat.
*   **±0.60 hingga ±0.79**: Korelasi kuat.
*   **±0.80 hingga ±1.0**: Korelasi sangat kuat.

**Poin Penting untuk Diingat:**

*   **Korelasi ≠ Kausalitas**: Korelasi hanya menunjukkan adanya hubungan atau pola antara dua variabel, bukan berarti satu variabel menyebabkan perubahan pada variabel lainnya. Mungkin ada faktor lain (variabel pengganggu) yang memengaruhi keduanya.
*   **Hubungan Linier**: Korelasi Pearson hanya mengukur hubungan *linier*. Jika hubungan antar variabel bersifat non-linier (misalnya, berbentuk kurva U), korelasi Pearson mungkin mendekati nol meskipun ada hubungan yang kuat.
*   **Sensitif terhadap Pencilan**: Nilai ekstrem (outlier) dapat secara signifikan memengaruhi nilai koefisien korelasi Pearson. Oleh karena itu, penting untuk selalu memvisualisasikan data (misalnya dengan scatter plot) selain menghitung korelasinya.

Dalam analisis data, memahami korelasi antar variabel adalah langkah krusial untuk mengidentifikasi variabel-variabel yang relevan, mendeteksi potensi masalah multikolinearitas (jika dua variabel independen sangat berkorelasi), dan mendapatkan wawasan awal tentang hubungan data sebelum membangun model.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Memastikan df sudah dimuat. Jika belum, uncomment baris di bawah:
# df = pd.read_csv("/content/sample_data/california_housing_train.csv")

print("Melakukan Analisis Korelasi Pearson untuk Kolom Numerik...")

# Pilih hanya kolom numerik untuk analisis korelasi
# df.select_dtypes(include=np.number) akan memilih semua kolom numerik
correlation_matrix = df.corr(method='pearson')

print("Matriks Korelasi Pearson:")
print(correlation_matrix)

# Visualisasi Heatmap Korelasi
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Heatmap Korelasi Pearson Antar Variabel Numerik', fontsize=16)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Analisis Korelasi Spearman: Pemahaman dan Interpretasi

**Korelasi Spearman (Spearman's ρ atau rho)** adalah ukuran statistik non-parametrik yang mengevaluasi seberapa kuat dan dalam arah apa hubungan monoton (monotonik) antara dua variabel. Berbeda dengan Korelasi Pearson yang mengukur hubungan linier, Korelasi Spearman bekerja dengan *peringkat* data, bukan nilai mentah, sehingga lebih fleksibel dalam menangkap berbagai jenis hubungan monoton.

**Kapan menggunakan Spearman daripada Pearson?**
*   Ketika asumsi normalitas data tidak terpenuhi (data tidak berdistribusi normal).
*   Ketika hubungan antar variabel adalah monotonik tetapi tidak harus linier (misalnya, salah satu variabel terus meningkat seiring yang lain meningkat, tetapi tidak dengan laju yang konstan).
*   Ketika ada nilai pencilan (outlier) yang signifikan, karena Spearman yang bekerja dengan peringkat menjadi lebih robust (tidak mudah terpengaruh) terhadap outlier dibandingkan Pearson.

**Rentang Nilai Koefisien Korelasi Spearman:**
Sama seperti Pearson, koefisien korelasi Spearman juga memiliki nilai antara -1 dan +1:

*   **ρ = +1**: Menunjukkan hubungan monoton positif sempurna. Peringkat satu variabel persis sama dengan peringkat variabel lain.
*   **ρ = -1**: Menunjukkan hubungan monoton negatif sempurna. Peringkat satu variabel berlawanan dengan peringkat variabel lain.
*   **ρ = 0**: Menunjukkan tidak ada hubungan monoton antara kedua variabel.

**Interpretasi Kekuatan Korelasi (Panduan Umum):**
Interpretasi kekuatan (sangat lemah, lemah, moderat, kuat, sangat kuat) umumnya sama dengan Pearson, namun perlu diingat bahwa ini mengacu pada hubungan *monotonik*, bukan hanya *linier*.

*   **0.0 hingga ±0.19**: Korelasi sangat lemah (dapat diabaikan).
*   **±0.20 hingga ±0.39**: Korelasi lemah.
*   **±0.40 hingga ±0.59**: Korelasi moderat.
*   **±0.60 hingga ±0.79**: Korelasi kuat.
*   **±0.80 hingga ±1.0**: Korelasi sangat kuat.

**Poin Penting untuk Diingat:**

*   **Korelasi ≠ Kausalitas**: Sama seperti Pearson, Spearman juga tidak menunjukkan hubungan sebab-akibat.
*   **Hubungan Monotonik**: Spearman efektif untuk mengukur hubungan di mana variabel cenderung bergerak bersama dalam arah yang sama (positif) atau berlawanan (negatif), meskipun laju perubahannya tidak konstan.
*   **Lebih Robust terhadap Outlier**: Karena menggunakan peringkat, Spearman kurang sensitif terhadap pencilan dibandingkan Pearson, menjadikannya pilihan yang baik untuk data yang mungkin mengandung nilai ekstrem.

Dengan membandingkan hasil Korelasi Pearson dan Spearman, kita bisa mendapatkan wawasan lebih dalam tentang sifat hubungan antar variabel.
